````python
# ============================================================
# ANALISIS DATA MASW UNTUK KARAKTERISASI TANAH DANGKAL
# Data simulasi | Bukan data pengukuran lapangan
# ============================================================
# Instalasi jika diperlukan:
# %pip install numpy pandas matplotlib scipy
#
# Folder data:
# D:\KULIAH\SEMESTER 7\KB PART 2\sample_data_MASW_simulasi
#
# File yang diperlukan:
# - synthetic_masw_seismic.csv
# - metadata.json
# - synthetic_reference_dispersion.csv
#
# CATATAN:
# Citra f-k adalah analisis awal. Inversi di bawah merupakan
# demonstrasi numerik sederhana, bukan pengganti forward modeling
# dispersi Rayleigh yang tervalidasi untuk studi geoteknik.
# ============================================================

from pathlib import Path
import json
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.signal import detrend, butter, sosfiltfilt, windows
from scipy.ndimage import gaussian_filter
from scipy.optimize import least_squares

# ------------------------------------------------------------
# 1. PENGATURAN FOLDER DAN OUTPUT
# ------------------------------------------------------------
DATA_DIR = Path(
    r"D:\KULIAH\SEMESTER 7\KB PART 2\sample_data_MASW_simulasi"
)
OUTPUT_DIR = DATA_DIR / "output_analisis_MASW"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

SEISMIC_FILE = DATA_DIR / "synthetic_masw_seismic.csv"
METADATA_FILE = DATA_DIR / "metadata.json"
DISPERSION_FILE = DATA_DIR / "synthetic_reference_dispersion.csv"

# Parameter preprocessing; sesuaikan bila data berbeda
APPLY_BANDPASS = False
FREQ_LOW_HZ = 5.0
FREQ_HIGH_HZ = 80.0

# Parameter citra f-k
F_MIN_HZ = 5.0
F_MAX_HZ = 45.0
C_MIN_M_S = 80.0
C_MAX_M_S = 500.0
N_C_VALUES = 220

# Parameter model inversi demonstrasi
N_LAYERS = 4
MAX_INVERSION_EVALUATIONS = 120

# ------------------------------------------------------------
# 2. PEMERIKSAAN DAN PEMBACAAN FILE
# ------------------------------------------------------------
required_files = [SEISMIC_FILE, METADATA_FILE, DISPERSION_FILE]
missing = [str(p) for p in required_files if not p.exists()]
if missing:
    raise FileNotFoundError(
        "File berikut tidak ditemukan:\n- " + "\n- ".join(missing) +
        "\nPastikan ZIP sudah diekstrak langsung ke folder DATA_DIR."
    )

seismic_df = pd.read_csv(SEISMIC_FILE)
with open(METADATA_FILE, "r", encoding="utf-8") as f:
    metadata = json.load(f)
dispersion_ref_df = pd.read_csv(DISPERSION_FILE)

if "time_s" not in seismic_df.columns:
    raise ValueError("CSV seismik harus memiliki kolom 'time_s'.")

channel_columns = [c for c in seismic_df.columns if c.startswith("CH")]
if len(channel_columns) < 4:
    raise ValueError("Channel CH tidak cukup untuk analisis multichannel.")

time_s = seismic_df["time_s"].to_numpy(dtype=float)
raw_data = seismic_df[channel_columns].to_numpy(dtype=float)
n_samples, n_channels = raw_data.shape

if n_samples < 16 or not np.all(np.diff(time_s) > 0):
    raise ValueError("Sumbu waktu tidak valid atau jumlah sampel terlalu sedikit.")

dt = float(metadata.get("sampling_interval_s", np.median(np.diff(time_s))))
fs = 1.0 / dt
dx = float(metadata["geophone_spacing_m"])
source_offset = float(metadata.get("source_offset_m", 0.0))
receiver_positions = np.asarray(
    metadata.get("receiver_positions_m", source_offset + np.arange(n_channels) * dx),
    dtype=float
)

if len(receiver_positions) != n_channels:
    raise ValueError("Jumlah posisi receiver tidak sama dengan jumlah channel.")
if not np.isclose(np.median(np.diff(time_s)), dt, rtol=0.02, atol=1e-8):
    warnings.warn("Interval waktu dari metadata berbeda dari interval waktu CSV.")

print("=" * 65)
print("RINGKASAN DATA MASW")
print("=" * 65)
print("Jenis data              :", metadata.get("dataset_type", "Tidak dinyatakan"))
print("Jumlah sampel           :", n_samples)
print("Jumlah channel          :", n_channels)
print("Interval sampling       :", f"{dt:.6f} s")
print("Frekuensi sampling      :", f"{fs:.2f} Hz")
print("Durasi rekaman          :", f"{time_s[-1] - time_s[0]:.3f} s")
print("Spasi antar-geophone    :", f"{dx:.2f} m")
print("Offset sumber           :", f"{source_offset:.2f} m")
print("Posisi receiver         :", receiver_positions)
print("Maksimum amplitudo awal :", f"{np.max(np.abs(raw_data)):.4f}")

# ------------------------------------------------------------
# 3. PREPROCESSING DASAR
# ------------------------------------------------------------
# Hilangkan mean dan tren linier per channel.
data = detrend(raw_data, axis=0, type="linear")

# Opsional: filter band-pass. Tidak diterapkan secara default.
if APPLY_BANDPASS:
    nyquist = fs / 2.0
    low = max(FREQ_LOW_HZ / nyquist, 1e-5)
    high = min(FREQ_HIGH_HZ / nyquist, 0.999)
    if low >= high:
        raise ValueError("Rentang band-pass tidak valid untuk sampling rate.")
    sos = butter(4, [low, high], btype="bandpass", output="sos")
    data = sosfiltfilt(sos, data, axis=0)
    print(f"Band-pass diterapkan: {FREQ_LOW_HZ}-{FREQ_HIGH_HZ} Hz")
else:
    print("Band-pass tidak diterapkan; data hanya detrend.")

# Normalisasi per trace untuk visualisasi (tidak mengubah data sumber).
trace_max = np.max(np.abs(data), axis=0)
trace_max[trace_max == 0] = 1.0
data_normalized = data / trace_max

# ------------------------------------------------------------
# 4. VISUALISASI REKAMAN SEISMIK MULTICHANNEL
# ------------------------------------------------------------
fig, ax = plt.subplots(figsize=(12, 7))
extent = [0.5, n_channels + 0.5, time_s[-1], time_s[0]]
im = ax.imshow(
    data_normalized,
    aspect="auto",
    cmap="gray",
    extent=extent,
    interpolation="nearest",
    vmin=-1,
    vmax=1
)
ax.set_xlabel("Channel Geophone")
ax.set_ylabel("Waktu (s)")
ax.set_title("Rekaman Seismik MASW Multichannel — Data Simulasi")
fig.colorbar(im, ax=ax, label="Amplitudo ternormalisasi")
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "01_rekaman_seismik.png", dpi=200)
plt.show()

# Wiggle plot untuk memeriksa bentuk trace
fig, ax = plt.subplots(figsize=(12, 7))
wiggle_scale = 0.45
for j in range(n_channels):
    tr = data_normalized[:, j]
    ax.plot(j + 1 + wiggle_scale * tr, time_s, color="black", linewidth=0.55)
ax.invert_yaxis()
ax.set_xlabel("Channel (offset wiggle)")
ax.set_ylabel("Waktu (s)")
ax.set_title("Wiggle Plot Rekaman MASW — Data Simulasi")
ax.grid(alpha=0.2)
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "02_wiggle_plot.png", dpi=200)
plt.show()

# ------------------------------------------------------------
# 5. CITRA DISPERSI AWAL MENGGUNAKAN TRANSFORMASI f-k
# ------------------------------------------------------------
# Catatan: metode f-k menghasilkan spektrum energi terhadap
# frekuensi dan bilangan gelombang. Konversi c=f/k memerlukan
# pemeriksaan konvensi bilangan gelombang dan aliasing spasial.
# Spasi geophone membatasi bilangan gelombang Nyquist.

# Hilangkan mean antar-channel dan gunakan taper spasial.
fk_input = data - np.mean(data, axis=1, keepdims=True)
space_taper = windows.tukey(n_channels, alpha=0.25)
time_taper = windows.tukey(n_samples, alpha=0.05)
fk_input = fk_input * time_taper[:, None] * space_taper[None, :]

# FFT waktu lalu FFT spasial
spec_t = np.fft.rfft(fk_input, axis=0)
freqs = np.fft.rfftfreq(n_samples, d=dt)
spec_fk = np.fft.fftshift(np.fft.fft(spec_t, axis=1), axes=1)
k_cycles_m = np.fft.fftshift(np.fft.fftfreq(n_channels, d=dx))

freq_mask = (freqs >= F_MIN_HZ) & (freqs <= min(F_MAX_HZ, fs / 2))
# Abaikan bilangan gelombang nol karena c=f/k tidak terdefinisi.
k_mask = np.abs(k_cycles_m) > 1e-10
freq_use = freqs[freq_mask]
k_use = k_cycles_m[k_mask]
energy_fk = np.abs(spec_fk[np.ix_(freq_mask, k_mask)]) ** 2

# Konversi kecepatan fase dari frekuensi / |k|.
# k di sini dalam cycles/m, sehingga c=f/|k|.
c_grid = np.linspace(C_MIN_M_S, C_MAX_M_S, N_C_VALUES)
disp_image = np.zeros((len(c_grid), len(freq_use)), dtype=float)

for i, f in enumerate(freq_use):
    c_values = f / np.abs(k_use)
    for ic, c in enumerate(c_grid):
        # Ambil energi di sekitar kecepatan yang sesuai.
        tolerance = max((C_MAX_M_S - C_MIN_M_S) / N_C_VALUES, 8.0)
        mask_c = np.abs(c_values - c) <= tolerance
        if np.any(mask_c):
            disp_image[ic, i] = np.sum(energy_fk[i, mask_c])

disp_image = gaussian_filter(disp_image, sigma=(1.0, 1.0))
if np.max(disp_image) > 0:
    disp_image_db = 10 * np.log10(disp_image / np.max(disp_image) + 1e-12)
else:
    disp_image_db = disp_image

fig, ax = plt.subplots(figsize=(11, 7))
mesh = ax.pcolormesh(
    freq_use, c_grid, disp_image_db,
    shading="auto", cmap="turbo", vmin=-35, vmax=0
)
ax.set_xlabel("Frekuensi (Hz)")
ax.set_ylabel("Kecepatan fase semu (m/s)")
ax.set_title("Citra Energi Dispersi f-k — Data Simulasi")
fig.colorbar(mesh, ax=ax, label="Energi relatif (dB)")
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "03_citra_dispersi_fk.png", dpi=200)
plt.show()

# Picking otomatis awal: kecepatan dengan energi maksimum tiap frekuensi.
# Picking ini harus diperiksa manual terhadap citra energi.
pick_idx = np.argmax(disp_image, axis=0)
picked_c = c_grid[pick_idx]
picked_energy = disp_image[pick_idx, np.arange(len(freq_use))]
valid_pick = (
    (freq_use >= F_MIN_HZ) &
    (freq_use <= F_MAX_HZ) &
    np.isfinite(picked_c) &
    (picked_energy > 0)
)
picked_df = pd.DataFrame({
    "frequency_Hz": freq_use[valid_pick],
    "phase_velocity_m_s": picked_c[valid_pick],
    "relative_energy": picked_energy[valid_pick],
    "pick_type": "AUTOMATIC_INITIAL_PICK_REQUIRES_QC"
})
picked_df.to_csv(OUTPUT_DIR / "picked_dispersion_initial.csv", index=False)

fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(
    picked_df["frequency_Hz"],
    picked_df["phase_velocity_m_s"],
    "o-", markersize=3, linewidth=1,
    label="Picking otomatis awal"
)
ax.set_xlabel("Frekuensi (Hz)")
ax.set_ylabel("Kecepatan fase semu (m/s)")
ax.set_title("Picking Awal Kurva Dispersi — Wajib QC")
ax.grid(alpha=0.3)
ax.legend()
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "04_picking_awal.png", dpi=200)
plt.show()

# ------------------------------------------------------------
# 6. KURVA DISPERSI REFERENSI SIMULASI
# ------------------------------------------------------------
# Kurva ini disediakan di paket data dan bukan hasil picking.
ref_f = dispersion_ref_df["frequency_Hz"].to_numpy(dtype=float)
ref_c = dispersion_ref_df["phase_velocity_m_s"].to_numpy(dtype=float)

fig, ax = plt.subplots(figsize=(9, 6))
ax.plot(ref_f, ref_c, "o-", label="Kurva referensi buatan")
ax.set_xlabel("Frekuensi (Hz)")
ax.set_ylabel("Kecepatan fase (m/s)")
ax.set_title("Kurva Dispersi Referensi — Bukan Data Observasi")
ax.grid(alpha=0.3)
ax.legend()
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "05_kurva_referensi.png", dpi=200)
plt.show()

# ------------------------------------------------------------
# 7. INVERSI Vs SEDERHANA (DEMONSTRASI)
# ------------------------------------------------------------
# Model berlapis dengan ketebalan lapisan pertama sampai N-1,
# dan lapisan terakhir half-space. Forward model berikut
# menggunakan kecepatan rata-rata berbobot sensitivitas kedalaman
# sebagai pendekatan demonstrasi. Ini BUKAN solver Rayleigh
# dispersion penuh dan hasilnya tidak boleh digunakan sebagai
# estimasi geoteknik final.
#
# Agar tidak menampilkan hasil palsu sebagai hasil observasi,
# inversi menggunakan kurva referensi simulasi yang tersedia.
# Jika ingin menggunakan picked_df, aktifkan USE_AUTOMATIC_PICK
# setelah picking dari citra dispersi diperiksa manual.

USE_AUTOMATIC_PICK = False

if USE_AUTOMATIC_PICK and len(picked_df) >= 8:
    inv_f = picked_df["frequency_Hz"].to_numpy()
    inv_c = picked_df["phase_velocity_m_s"].to_numpy()
    inversion_source = "picking otomatis awal (belum tervalidasi)"
else:
    inv_f = ref_f.copy()
    inv_c = ref_c.copy()
    inversion_source = "kurva referensi simulasi (bukan observasi)"

# Pilih subset frekuensi untuk stabilitas dan kecepatan komputasi.
valid = (
    np.isfinite(inv_f) & np.isfinite(inv_c) &
    (inv_f > 0) & (inv_c > 0)
)
inv_f, inv_c = inv_f[valid], inv_c[valid]
if len(inv_f) < 6:
    raise ValueError("Titik dispersi valid tidak cukup untuk demonstrasi inversi.")

order = np.argsort(inv_f)
inv_f, inv_c = inv_f[order], inv_c[order]

# Lapisan model: ketebalan dan Vs. Kedalaman investigasi adalah asumsi
# model untuk demonstrasi, bukan fakta yang diukur.
layer_thickness = np.array([3.0, 5.0, 8.0])  # m; 4 lapisan termasuk half-space
depth_boundaries = np.r_[0.0, np.cumsum(layer_thickness)]
layer_midpoints = np.r_[
    layer_thickness[0] / 2,
    layer_thickness[0] + layer_thickness[1] / 2,
    layer_thickness[0] + layer_thickness[1] + layer_thickness[2] / 2,
    np.sum(layer_thickness) + 5.0
]
layer_bottoms = np.r_[np.cumsum(layer_thickness), np.inf]

def sensitivity_weights(frequencies, mids):
    # Panjang gelombang referensi kasar; kedalaman sensitivitas ~ lambda/3.
    # Bobot eksponensial ini hanya pendekatan ilustratif.
    weights = []
    for f in frequencies:
        lam = max(200.0 / f, 1.0)
        zscale = max(lam / 3.0, 0.5)
        w = np.exp(-mids / zscale)
        w = w / np.sum(w)
        weights.append(w)
    return np.asarray(weights)

weights = sensitivity_weights(inv_f, layer_midpoints)

def forward_approx(vs):
    # Kecepatan fase aproksimatif; bukan solusi persamaan dispersi Rayleigh.
    return np.sum(weights * vs[None, :], axis=1) * 0.92

def residual(log_vs):
    vs = np.exp(log_vs)
    pred = forward_approx(vs)
    return (pred - inv_c) / max(np.median(inv_c), 1.0)

initial_vs = np.array([180.0, 250.0, 350.0, 450.0])
lower_vs = np.array([80.0, 80.0, 80.0, 80.0])
upper_vs = np.array([900.0, 900.0, 1200.0, 1500.0])

result = least_squares(
    residual,
    np.log(initial_vs),
    bounds=(np.log(lower_vs), np.log(upper_vs)),
    max_nfev=MAX_INVERSION_EVALUATIONS
)
vs_model = np.exp(result.x)
c_pred = forward_approx(vs_model)
rmse_approx = float(np.sqrt(np.mean((c_pred - inv_c) ** 2)))

model_df = pd.DataFrame({
    "layer": np.arange(1, N_LAYERS + 1),
    "top_depth_m": [0.0, 3.0, 8.0, 16.0],
    "bottom_depth_m": [3.0, 8.0, 16.0, np.nan],
    "Vs_m_s": vs_model,
    "model_status": "DEMONSTRATION_APPROXIMATION_NOT_VALIDATED"
})
model_df.to_csv(OUTPUT_DIR / "model_Vs_demonstrasi.csv", index=False)

fig, ax = plt.subplots(figsize=(6, 8))
top = 0.0
for i, vs in enumerate(vs_model):
    bottom = layer_bottoms[i]
    if not np.isfinite(bottom):
        bottom = 25.0  # hanya batas tampilan half-space
    ax.fill_betweenx([top, bottom], 0, vs, alpha=0.35)
    ax.plot([vs, vs], [top, bottom], linewidth=2)
    ax.text(vs + 8, (top + bottom) / 2, f"Vs={vs:.0f} m/s", va="center")
    top = bottom
ax.set_ylim(25, 0)
ax.set_xlim(0, max(vs_model) * 1.35)
ax.set_xlabel("Vs (m/s)")
ax.set_ylabel("Kedalaman (m)")
ax.set_title("Model Vs Berlapis — Demonstrasi Aproksimatif")
ax.grid(alpha=0.25)
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "06_model_Vs_demonstrasi.png", dpi=200)
plt.show()

fig, ax = plt.subplots(figsize=(9, 6))
ax.plot(inv_f, inv_c, "o", label="Kurva input inversi")
ax.plot(inv_f, c_pred, "-", label="Prediksi model aproksimatif")
ax.set_xlabel("Frekuensi (Hz)")
ax.set_ylabel("Kecepatan fase (m/s)")
ax.set_title("Pemeriksaan Kecocokan Inversi Demonstrasi")
ax.grid(alpha=0.3)
ax.legend()
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "07_kecocokan_inversi_demonstrasi.png", dpi=200)
plt.show()

# ------------------------------------------------------------
# 8. PERHITUNGAN Vs30
# ------------------------------------------------------------
# Vs30 = 30 / sum(h_i / Vs_i), untuk total ketebalan 30 m.
# Model demonstrasi di atas hanya memiliki ketebalan berhingga 16 m
# dan half-space tidak memiliki batas kedalaman yang terdefinisi.
# Karena itu Vs30 TIDAK dihitung sebagai hasil yang sah dari model ini.
finite_depth = float(np.sum(layer_thickness))
if finite_depth >= 30.0:
    thickness_for_vs30 = layer_thickness.copy()
    vs_for_vs30 = vs_model[:len(layer_thickness)]
    if np.isclose(finite_depth, 30.0):
        vs30 = 30.0 / np.sum(thickness_for_vs30 / vs_for_vs30)
        vs30_status = "Dihitung dari model berhingga 30 m"
    else:
        vs30 = None
        vs30_status = "Perlu memotong lapisan pada kedalaman tepat 30 m"
else:
    vs30 = None
    vs30_status = (
        f"TIDAK DIHITUNG: model berhingga hanya {finite_depth:.1f} m; "
        "Vs half-space tidak dapat diasumsikan untuk seluruh kedalaman 30 m."
    )

print("\n" + "=" * 65)
print("HASIL ANALISIS")
print("=" * 65)
print("Sumber kurva inversi:", inversion_source)
print("Status optimizer:", result.message)
print("RMSE aproksimatif:", f"{rmse_approx:.2f} m/s")
print("\nModel Vs demonstrasi:")
print(model_df.to_string(index=False))
print("\nVs30:", vs30 if vs30 is not None else "Tidak dihitung")
print("Status Vs30:", vs30_status)

# ------------------------------------------------------------
# 9. SIMPAN RINGKASAN DAN KETERBATASAN
# ------------------------------------------------------------
summary = {
    "dataset_type": metadata.get("dataset_type", "unknown"),
    "n_channels": int(n_channels),
    "n_samples": int(n_samples),
    "sampling_interval_s": float(dt),
    "sampling_frequency_Hz": float(fs),
    "geophone_spacing_m": float(dx),
    "source_offset_m": float(source_offset),
    "dispersion_image_method": "2D temporal-spatial FFT (f-k), preliminary",
    "automatic_picking_status": "Initial pick only; requires manual quality control",
    "inversion_source": inversion_source,
    "inversion_method": "Layered approximate sensitivity-weighted demonstration; not full Rayleigh solver",
    "Vs_model_m_s": [float(v) for v in vs_model],
    "approximate_fit_RMSE_m_s": rmse_approx,
    "Vs30_m_s": None if vs30 is None else float(vs30),
    "Vs30_status": vs30_status,
    "limitations": [
        "All provided data are synthetic, not field measurements.",
        "The reference dispersion curve is constructed, not picked from measured dispersion image.",
        "The f-k dispersion image is preliminary and can be affected by spatial aliasing, limited array aperture, and synthetic wavefield construction.",
        "Automatic picking requires visual/manual quality control.",
        "The inversion is an approximate demonstration and is not a full Rayleigh-wave dispersion forward solver.",
        "The 30 m depth requirement is not met by the finite layer thicknesses in this demonstration model.",
        "Do not use the demonstration Vs model or inferred classes for engineering design."
    ]
}
with open(OUTPUT_DIR / "ringkasan_analisis.json", "w", encoding="utf-8") as f:
    json.dump(summary, f, indent=2, ensure_ascii=False)

with open(OUTPUT_DIR / "CATATAN_PENTING.txt", "w", encoding="utf-8") as f:
    f.write(
        "HASIL DEMONSTRASI MASW — DATA SIMULASI\n\n"
        "Semua data input adalah simulasi. Citra f-k dan picking otomatis merupakan\n"
        "hasil eksplorasi awal yang wajib diperiksa. Inversi Vs memakai forward model\n"
        "aproksimatif, bukan pemecah dispersi Rayleigh yang tervalidasi. Karena itu\n"
        "model Vs ini hanya demonstrasi kode, bukan interpretasi geoteknik final.\n\n"
        "Vs30 tidak dihitung karena model tidak menyediakan profil Vs yang memadai\n"
        "hingga kedalaman 30 m. Untuk hasil kuantitatif, gunakan solver dispersi\n"
        "Rayleigh yang sesuai, parameter Vp/densitas/Poisson yang beralasan, picking\n"
        "manual berkualitas, dan model kedalaman yang mencakup 30 m.\n"
    )

print("\nSemua output disimpan di:")
print(OUTPUT_DIR)
print("\nFile utama:")
for p in sorted(OUTPUT_DIR.iterdir()):
    print("-", p.name)

````

````python
# ============================================================
# ANALISIS DATA ERT UNTUK IDENTIFIKASI KONDISI BAWAH PERMUKAAN
# Konfigurasi: Wenner-Schlumberger | 24 elektroda | spasi 5 m
# DATA SIMULASI — BUKAN DATA PENGUKURAN LAPANGAN
# ============================================================
# Instalasi:
#   pip install numpy pandas matplotlib scipy
#
# Folder data:
#   D:\KULIAH\SEMESTER 7\KB PART 2\sample_data_ERT
#
# Catatan metodologis:
# - Pseudosection resistivitas semu bukan penampang resistivitas sebenarnya.
# - Bagian "model regularisasi aproksimatif" menggunakan sensitivitas
#   pedagogis yang disederhanakan. Ini BUKAN solver forward/inversi ERT
#   penuh berbasis finite-element/finite-difference.
# - Jangan menyebut hasil model aproksimatif sebagai inversi lapangan.
#   Untuk inversi ERT kuantitatif, gunakan solver tervalidasi seperti
#   pyGIMLi atau ResIPy dengan mesh, error model, dan QC yang tepat.
# ============================================================

from pathlib import Path
import json
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.interpolate import griddata
from scipy.ndimage import gaussian_filter
from scipy.sparse import diags, eye, kron
from scipy.sparse.linalg import lsqr

# ------------------------------------------------------------
# 1. PATH DAN PENGATURAN
# ------------------------------------------------------------
DATA_DIR = Path(r"D:\KULIAH\SEMESTER 7\KB PART 2\sample_data_ERT")
OUTPUT_DIR = DATA_DIR / "output_analisis_ERT"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

DATA_FILE = DATA_DIR / "synthetic_ert_wenner_schlumberger.csv"
METADATA_FILE = DATA_DIR / "metadata.json"

# Pengaturan QC
MIN_CURRENT_MA = 0.01
MIN_ABS_VOLTAGE_MV = 1e-8
ROBUST_Z_THRESHOLD = 3.5

# Pengaturan model regularisasi aproksimatif
NX_MODEL = 36
NZ_MODEL = 16
MODEL_DEPTH_M = 35.0
LAMBDA_SMOOTH = 2.0
LAMBDA_DAMPING = 0.15

# ------------------------------------------------------------
# 2. MEMBACA DAN MEMERIKSA DATA
# ------------------------------------------------------------
for p in [DATA_FILE, METADATA_FILE]:
    if not p.exists():
        raise FileNotFoundError(
            f"File tidak ditemukan: {p}\n"
            "Pastikan paket ZIP telah diekstrak ke DATA_DIR."
        )

df = pd.read_csv(DATA_FILE)
with open(METADATA_FILE, "r", encoding="utf-8") as f:
    metadata = json.load(f)

required_columns = [
    "A_x_m", "M_x_m", "N_x_m", "B_x_m",
    "current_mA", "voltage_mV", "K_ohm_m"
]
missing = [c for c in required_columns if c not in df.columns]
if missing:
    raise ValueError(f"Kolom wajib tidak ditemukan: {missing}")

# Hitung ulang resistivitas semu dari data arus, tegangan, dan K.
# Karena mV/mA = ohm, rhoa = K * (V_mV / I_mA) menghasilkan ohm-m.
df["rhoa_recalculated_ohm_m"] = (
    df["K_ohm_m"] * df["voltage_mV"] / df["current_mA"]
)

# Catat perbedaan dengan nilai tersimpan, bila tersedia.
if "rhoa_ohm_m" in df.columns:
    df["rhoa_difference_pct"] = (
        100.0 * (df["rhoa_recalculated_ohm_m"] - df["rhoa_ohm_m"])
        / df["rhoa_ohm_m"].replace(0, np.nan)
    )
else:
    df["rhoa_difference_pct"] = np.nan

# ------------------------------------------------------------
# 3. QUALITY CONTROL (QC)
# ------------------------------------------------------------
df["qc_invalid_current"] = (
    ~np.isfinite(df["current_mA"]) |
    (df["current_mA"] <= MIN_CURRENT_MA)
)
df["qc_invalid_voltage"] = (
    ~np.isfinite(df["voltage_mV"]) |
    (np.abs(df["voltage_mV"]) <= MIN_ABS_VOLTAGE_MV)
)
df["qc_nonpositive_rhoa"] = (
    ~np.isfinite(df["rhoa_recalculated_ohm_m"]) |
    (df["rhoa_recalculated_ohm_m"] <= 0)
)

# Robust z-score pada log rhoa. Ini flag statistik, bukan bukti otomatis
# bahwa data harus dibuang; perlu diperiksa terhadap geometri dan data lapangan.
valid_for_stats = (
    np.isfinite(df["rhoa_recalculated_ohm_m"]) &
    (df["rhoa_recalculated_ohm_m"] > 0)
)
log_rhoa = np.log(df.loc[valid_for_stats, "rhoa_recalculated_ohm_m"])
median_log = float(log_rhoa.median()) if len(log_rhoa) else 0.0
mad_log = float(np.median(np.abs(log_rhoa - median_log))) if len(log_rhoa) else 0.0
robust_z = pd.Series(np.nan, index=df.index, dtype=float)
if mad_log > 0:
    robust_z.loc[valid_for_stats] = (
        0.6745 * (log_rhoa - median_log) / mad_log
    )
df["qc_robust_z_log_rhoa"] = robust_z
df["qc_statistical_outlier"] = (
    df["qc_robust_z_log_rhoa"].abs() > ROBUST_Z_THRESHOLD
)

# Pertahankan flag outlier buatan dari dataset sebagai informasi tambahan.
if "synthetic_outlier_injected" in df.columns:
    df["qc_injected_demo_outlier"] = df["synthetic_outlier_injected"].astype(bool)
else:
    df["qc_injected_demo_outlier"] = False

df["qc_reject_for_plot"] = (
    df["qc_invalid_current"] |
    df["qc_invalid_voltage"] |
    df["qc_nonpositive_rhoa"]
)
df["qc_flag_any"] = (
    df["qc_reject_for_plot"] |
    df["qc_statistical_outlier"] |
    df["qc_injected_demo_outlier"]
)

# Simpan seluruh data dengan kolom QC
df.to_csv(OUTPUT_DIR / "01_data_dengan_QC.csv", index=False)

# Data untuk plotting/inversi: buang hanya data invalid secara dasar.
# Outlier statistik tidak otomatis dihapus; dibuat pilihan konservatif.
clean = df.loc[~df["qc_reject_for_plot"]].copy()
# Untuk demonstrasi, eksklusi flag outlier buatan dan outlier statistik
# dari model aproksimatif, tetapi simpan semuanya di file QC.
fit_data = clean.loc[
    (~clean["qc_statistical_outlier"]) &
    (~clean["qc_injected_demo_outlier"])
].copy()
if len(fit_data) < 10:
    warnings.warn("Data valid untuk model sedikit; gunakan hasil hanya sebagai demonstrasi.")
    fit_data = clean.copy()

# ------------------------------------------------------------
# 4. RINGKASAN DATA DAN CONTOH HITUNGAN
# ------------------------------------------------------------
print("=" * 70)
print("ANALISIS ERT — DATA SIMULASI")
print("=" * 70)
print("Jenis data             :", metadata.get("dataset_type"))
print("Konfigurasi            :", metadata.get("configuration"))
print("Jumlah elektroda       :", metadata.get("number_of_electrodes"))
print("Spasi elektroda        :", metadata.get("electrode_spacing_m"), "m")
print("Panjang lintasan       :", metadata.get("line_length_m"), "m")
print("Jumlah quadrupole      :", len(df))
print("Data valid dasar       :", len(clean))
print("Flag outlier statistik :", int(df["qc_statistical_outlier"].sum()))
print("Outlier simulasi       :", int(df["qc_injected_demo_outlier"].sum()))
print("Data digunakan model   :", len(fit_data))

example = df.iloc[0]
print("\nContoh perhitungan rhoa:")
print(f"K = {example['K_ohm_m']:.3f} m")
print(f"Delta V = {example['voltage_mV']:.6f} mV")
print(f"I = {example['current_mA']:.4f} mA")
print(
    "rhoa = K * (DeltaV/I) = "
    f"{example['K_ohm_m']:.3f} * "
    f"({example['voltage_mV']:.6f}/{example['current_mA']:.4f}) "
    f"= {example['rhoa_recalculated_ohm_m']:.3f} ohm m"
)

# ------------------------------------------------------------
# 5. GRAFIK KUALITAS DATA
# ------------------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
axes[0].hist(
    np.log10(clean["rhoa_recalculated_ohm_m"]),
    bins=24, edgecolor="black"
)
axes[0].set_xlabel("log10 Resistivitas Semu (ohm m)")
axes[0].set_ylabel("Jumlah data")
axes[0].set_title("Distribusi Resistivitas Semu")

colors = np.where(df["qc_flag_any"], "tab:red", "tab:blue")
axes[1].scatter(
    (df["A_x_m"] + df["B_x_m"]) / 2,
    df["rhoa_recalculated_ohm_m"],
    c=colors, s=22
)
axes[1].set_yscale("log")
axes[1].set_xlabel("Titik tengah array (m)")
axes[1].set_ylabel("Resistivitas semu (ohm m)")
axes[1].set_title("QC: biru = tidak ter-flag, merah = ter-flag")
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "02_grafik_quality_control.png", dpi=200)
plt.show()

# ------------------------------------------------------------
# 6. PSEUDOSECTION RESISTIVITAS SEMU
# ------------------------------------------------------------
# Koordinat pseudo-depth berikut adalah pendekatan plotting berbasis
# panjang array; BUKAN kedalaman aktual yang terukur.
# Dipakai pseudo-depth = 0.19 * AB sebagai konvensi visual sederhana.
plot_df = fit_data.copy()
plot_df["x_mid_m"] = (plot_df["A_x_m"] + plot_df["B_x_m"]) / 2.0
plot_df["pseudo_depth_m"] = 0.19 * plot_df["AB_m"]

fig, ax = plt.subplots(figsize=(13, 6))
sc = ax.scatter(
    plot_df["x_mid_m"],
    plot_df["pseudo_depth_m"],
    c=plot_df["rhoa_recalculated_ohm_m"],
    s=42,
    cmap="turbo",
    norm=plt.matplotlib.colors.LogNorm(
        vmin=max(plot_df["rhoa_recalculated_ohm_m"].min(), 1e-3),
        vmax=plot_df["rhoa_recalculated_ohm_m"].max()
    ),
    edgecolors="black",
    linewidths=0.15
)
ax.set_xlabel("Jarak sepanjang lintasan (m)")
ax.set_ylabel("Pseudo-depth (m)")
ax.set_title("Pseudosection Resistivitas Semu — Bukan Penampang Inversi")
ax.invert_yaxis()
cb = fig.colorbar(sc, ax=ax)
cb.set_label("Resistivitas semu (ohm m)")
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "03_pseudosection_resistivitas_semu.png", dpi=200)
plt.show()

# ------------------------------------------------------------
# 7. MODEL 2D REGULARISASI APROKSIMATIF (BUKAN SOLVER ERT PENUH)
# ------------------------------------------------------------
# Untuk menghindari klaim berlebihan, model ini diberi label
# "regularized approximate model". Operator sensitivitas merupakan
# kernel sederhana berdasarkan jarak horizontal dan pseudo-depth.
# Ini berguna untuk demonstrasi regularisasi dan bukan forward physics
# penuh. Tidak boleh diperlakukan sebagai inversi ERT kuantitatif.

x_nodes = np.linspace(0.0, metadata["line_length_m"], NX_MODEL)
z_nodes = np.linspace(1.0, MODEL_DEPTH_M, NZ_MODEL)
XX, ZZ = np.meshgrid(x_nodes, z_nodes)
cell_x = XX.ravel()
cell_z = ZZ.ravel()
n_cells = len(cell_x)

obs_x = plot_df["x_mid_m"].to_numpy()
obs_z = plot_df["pseudo_depth_m"].to_numpy()
obs_logrho = np.log(plot_df["rhoa_recalculated_ohm_m"].to_numpy())

# Kernel: horizontal gaussian spread + depth decay, normalisasi setiap baris.
# Sensitivitas depth dibangun hanya sebagai pendekatan pedagogis.
A = np.empty((len(plot_df), n_cells), dtype=float)
for i, (xo, zo) in enumerate(zip(obs_x, obs_z)):
    sx = max(7.5, 0.45 * (zo + 5.0))
    sz = max(4.0, 0.65 * (zo + 2.0))
    kernel = np.exp(-0.5 * ((cell_x - xo) / sx) ** 2) * np.exp(
        -np.abs(cell_z - max(zo, 1.0)) / sz
    )
    total = kernel.sum()
    A[i, :] = kernel / total if total > 0 else np.ones(n_cells) / n_cells

# Bangun operator smoothness beda hingga horizontal dan vertikal.
def smoothness_matrix(nx, nz):
    # horizontal differences
    n = nx * nz
    rows, cols, vals = [], [], []
    r = 0
    for iz in range(nz):
        for ix in range(nx - 1):
            i1 = iz * nx + ix
            i2 = i1 + 1
            rows.extend([r, r]); cols.extend([i1, i2]); vals.extend([1.0, -1.0])
            r += 1
    for iz in range(nz - 1):
        for ix in range(nx):
            i1 = iz * nx + ix
            i2 = (iz + 1) * nx + ix
            rows.extend([r, r]); cols.extend([i1, i2]); vals.extend([1.0, -1.0])
            r += 1
    from scipy.sparse import csr_matrix
    return csr_matrix((vals, (rows, cols)), shape=(r, n))

L = smoothness_matrix(NX_MODEL, NZ_MODEL)
# Solve augmented least squares in log-resistivity:
# [A; sqrt(lambda)*L; sqrt(damping)*I] m = [d; 0; m0]
# Use lsqr on sparse augmented matrix.
from scipy.sparse import csr_matrix, vstack
A_sparse = csr_matrix(A)
I_sparse = eye(n_cells, format="csr")
L_weighted = np.sqrt(LAMBDA_SMOOTH) * L
I_weighted = np.sqrt(LAMBDA_DAMPING) * I_sparse
A_aug = vstack([A_sparse, L_weighted, I_weighted], format="csr")

rho_background = float(np.median(plot_df["rhoa_recalculated_ohm_m"]))
m0 = np.log(rho_background) * np.ones(n_cells)
rhs = np.r_[obs_logrho, np.zeros(L.shape[0]), np.sqrt(LAMBDA_DAMPING) * m0]
# For the smoothness block, target is zero difference; damping target is background.
solution = lsqr(A_aug, rhs, iter_lim=400, atol=1e-5, btol=1e-5)
model_logrho = solution[0]
model_rho = np.exp(model_logrho).reshape(NZ_MODEL, NX_MODEL)

# Approximate data fit and log-RMS. This is fit to the approximate kernel only.
pred_logrho = A @ model_logrho
rms_log = float(np.sqrt(np.mean((pred_logrho - obs_logrho) ** 2)))
rms_pct_approx = float(100 * np.sqrt(np.mean((np.exp(pred_logrho) - np.exp(obs_logrho))**2)) /
                       max(np.mean(np.exp(obs_logrho)), 1e-9))

fig, ax = plt.subplots(figsize=(13, 6))
mesh = ax.pcolormesh(
    x_nodes, z_nodes, model_rho,
    shading="auto", cmap="turbo",
    norm=plt.matplotlib.colors.LogNorm(
        vmin=max(float(model_rho.min()), 1e-3),
        vmax=float(model_rho.max())
    )
)
ax.set_xlabel("Jarak sepanjang lintasan (m)")
ax.set_ylabel("Kedalaman model (m)")
ax.set_title("Model Resistivitas 2D Regularisasi Aproksimatif — BUKAN Inversi ERT Penuh")
ax.invert_yaxis()
cb = fig.colorbar(mesh, ax=ax)
cb.set_label("Resistivitas model aproksimatif (ohm m)")
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "04_model_resistivitas_2D_aproksimatif.png", dpi=200)
plt.show()

# ------------------------------------------------------------
# 8. PLOT MODEL VS DATA OBSERVASI SEMU
# ------------------------------------------------------------
fig, ax = plt.subplots(figsize=(9, 5))
ax.scatter(
    np.exp(obs_logrho), np.exp(pred_logrho),
    s=24, alpha=0.75, edgecolors="black", linewidths=0.2
)
lims = [
    min(np.exp(obs_logrho).min(), np.exp(pred_logrho).min()),
    max(np.exp(obs_logrho).max(), np.exp(pred_logrho).max())
]
ax.plot(lims, lims, "k--", label="Garis 1:1")
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("Resistivitas semu input (ohm m)")
ax.set_ylabel("Respons prediksi kernel aproksimatif (ohm m)")
ax.set_title("Pemeriksaan Fit Model Aproksimatif (Bukan RMS Inversi ERT Standar)")
ax.grid(alpha=0.25, which="both")
ax.legend()
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "05_fit_model_aproksimatif.png", dpi=200)
plt.show()

# ------------------------------------------------------------
# 9. INTERPRETASI ZONA RESISTIVITAS (DESKRIPTIF, TIDAK UNIK)
# ------------------------------------------------------------
# Ambang ini dibuat relatif terhadap model sintetis, bukan standar litologi.
p33, p67 = np.percentile(model_rho, [33, 67])
zones = np.full(model_rho.shape, "Sedang", dtype=object)
zones[model_rho <= p33] = "Relatif rendah"
zones[model_rho >= p67] = "Relatif tinggi"

fig, ax = plt.subplots(figsize=(13, 6))
zone_code = np.zeros_like(model_rho, dtype=int)
zone_code[zones == "Relatif rendah"] = 0
zone_code[zones == "Sedang"] = 1
zone_code[zones == "Relatif tinggi"] = 2
from matplotlib.colors import ListedColormap
cmap = ListedColormap(["#2c7bb6", "#ffffbf", "#d7191c"])
im = ax.pcolormesh(x_nodes, z_nodes, zone_code, shading="auto", cmap=cmap, vmin=0, vmax=2)
ax.set_xlabel("Jarak sepanjang lintasan (m)")
ax.set_ylabel("Kedalaman model (m)")
ax.set_title("Zona Resistivitas Relatif — Interpretasi Awal Data Simulasi")
ax.invert_yaxis()
from matplotlib.patches import Patch
ax.legend(handles=[
    Patch(color="#2c7bb6", label=f"Relatif rendah (≤ {p33:.1f} Ωm)"),
    Patch(color="#ffffbf", label=f"Sedang ({p33:.1f}–{p67:.1f} Ωm)"),
    Patch(color="#d7191c", label=f"Relatif tinggi (≥ {p67:.1f} Ωm)")
], loc="upper right")
fig.tight_layout()
fig.savefig(OUTPUT_DIR / "06_zona_resistivitas_relief.png", dpi=200)
plt.show()

# ------------------------------------------------------------
# 10. SIMPAN OUTPUT TABEL DAN RINGKASAN
# ------------------------------------------------------------
model_rows = []
for iz, z in enumerate(z_nodes):
    for ix, x in enumerate(x_nodes):
        model_rows.append({
            "x_m": x,
            "depth_m": z,
            "resistivity_ohm_m": model_rho[iz, ix],
            "model_type": "REGULARIZED_APPROXIMATE_NOT_FULL_ERT_INVERSION"
        })
model_df = pd.DataFrame(model_rows)
model_df.to_csv(OUTPUT_DIR / "07_grid_model_resistivitas_aproksimatif.csv", index=False)

summary = {
    "dataset_type": metadata.get("dataset_type"),
    "configuration": metadata.get("configuration"),
    "number_of_electrodes": int(metadata.get("number_of_electrodes")),
    "electrode_spacing_m": float(metadata.get("electrode_spacing_m")),
    "line_length_m": float(metadata.get("line_length_m")),
    "number_of_measurements": int(len(df)),
    "number_basic_valid": int(len(clean)),
    "number_statistical_outliers": int(df["qc_statistical_outlier"].sum()),
    "number_injected_demo_outliers": int(df["qc_injected_demo_outlier"].sum()),
    "number_used_in_approximate_model": int(len(plot_df)),
    "apparent_resistivity_formula": "rhoa = K * (DeltaV/I); mV/mA = ohm",
    "geometric_factor_formula": metadata.get("geometric_factor_formula"),
    "model_type": "Regularized approximate sensitivity-kernel model; not full physics-based ERT inversion",
    "lambda_smooth": LAMBDA_SMOOTH,
    "lambda_damping": LAMBDA_DAMPING,
    "model_cells": int(n_cells),
    "approximate_log_RMS": rms_log,
    "approximate_relative_RMS_percent": rms_pct_approx,
    "model_resistivity_min_ohm_m": float(model_rho.min()),
    "model_resistivity_median_ohm_m": float(np.median(model_rho)),
    "model_resistivity_max_ohm_m": float(model_rho.max()),
    "relative_zone_thresholds_ohm_m": {"p33": float(p33), "p67": float(p67)},
    "limitations": [
        "All data are synthetic, not field measurements.",
        "Pseudodepth is a plotting convention and is not true depth.",
        "The regularized approximate model uses a simplified sensitivity kernel, not a full ERT forward solver.",
        "The reported fit statistics are for the approximate kernel and are not standard ERT inversion RMS.",
        "Resistivity alone cannot uniquely identify lithology, strength, or saturation.",
        "Geotechnical interpretation requires geological, borehole, SPT, groundwater, and/or laboratory constraints."
    ]
}
with open(OUTPUT_DIR / "08_ringkasan_analisis.json", "w", encoding="utf-8") as f:
    json.dump(summary, f, indent=2, ensure_ascii=False)

with open(OUTPUT_DIR / "CATATAN_METODOLOGI.txt", "w", encoding="utf-8") as f:
    f.write(
        "ANALISIS ERT — DATA SIMULASI\n\n"
        "Semua data adalah data simulasi. Resistivitas semu dihitung dengan rhoa=K*DeltaV/I.\n"
        "Pseudosection menggunakan pseudo-depth untuk visualisasi dan bukan kedalaman aktual.\n"
        "Model 2D yang dibuat adalah model regularisasi aproksimatif berbasis kernel sensitivitas\n"
        "sederhana, bukan inversi ERT penuh berbasis fisika. Statistik fit yang dicetak adalah\n"
        "statistik untuk model aproksimatif dan bukan RMS standar dari perangkat lunak inversi.\n\n"
        "Interpretasi resistivitas bersifat tidak unik. Resistivitas rendah dapat berkaitan dengan\n"
        "lempung, air/kejenuhan, fluida konduktif, atau alterasi; resistivitas tinggi dapat berkaitan\n"
        "dengan batuan kompak/kering, tetapi tidak boleh dipastikan tanpa data pendukung.\n"
        "Untuk inversi kuantitatif, gunakan solver seperti pyGIMLi/ResIPy dan lakukan QC, mesh,\n"
        "error model, pemilihan regularisasi, dan verifikasi model secara benar.\n"
    )

print("\nSelesai. Output disimpan di:")
print(OUTPUT_DIR)
print("\nFile output:")
for p in sorted(OUTPUT_DIR.iterdir()):
    print("-", p.name)
````

````python

# ================================================================
# PEMETAAN KERENTANAN LONGSOR BERBASIS GIS
# Mendukung Analisis Geoteknik
# Platform: Jupyter Notebook / Python
# ================================================================

from pathlib import Path
import json
import warnings

import numpy as np
import pandas as pd
import geopandas as gpd
import rasterio

from rasterio.features import rasterize
from rasterio.warp import reproject, Resampling
from scipy.ndimage import distance_transform_edt

import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap, BoundaryNorm

warnings.filterwarnings("ignore", category=RuntimeWarning)

# ================================================================
# 1. PENGATURAN DIREKTORI
# ================================================================

BASE_DIR = Path(
    r"D:\KULIAH\SEMESTER 7\KB PART 2\sample_data_GIS_Longsor"
)

DATA_DIR = BASE_DIR / "data"
OUT_DIR = BASE_DIR / "output"
OUT_DIR.mkdir(parents=True, exist_ok=True)

# Sistem koordinat proyeksi dalam meter.
# EPSG:32750 = WGS 84 / UTM zone 50S.
TARGET_CRS = "EPSG:32750"

# Bobot parameter kerentanan longsor.
WEIGHTS = {
    "slope": 0.35,
    "lithology": 0.20,
    "lulc": 0.15,
    "rainfall": 0.20,
    "river_distance": 0.10,
}

assert np.isclose(sum(WEIGHTS.values()), 1.0), \
    "Jumlah bobot harus sama dengan 1."

print("=" * 65)
print("PEMETAAN KERENTANAN LONGSOR BERBASIS GIS")
print("=" * 65)
print("Folder data   :", DATA_DIR)
print("Folder output :", OUT_DIR)

if not DATA_DIR.exists():
    raise FileNotFoundError(
        f"Folder data tidak ditemukan: {DATA_DIR}"
    )


# ================================================================
# 2. MENCARI FILE INPUT SECARA OTOMATIS
# ================================================================

def find_file(folder, keywords, extensions):
    """
    Mencari file berdasarkan kata kunci nama file.
    Jika ditemukan beberapa file, pilih nama yang paling cocok.
    """
    candidates = [
        p for p in folder.rglob("*")
        if p.is_file()
        and p.suffix.lower() in extensions
        and not any(
            part.lower() in {"output", "hasil", "archive", "backup"}
            for part in p.parts
        )
    ]

    matches = [
        p for p in candidates
        if any(k.lower() in p.stem.lower() for k in keywords)
    ]

    if not matches:
        return None

    matches.sort(
        key=lambda p: (
            -sum(k.lower() in p.stem.lower() for k in keywords),
            len(p.name)
        )
    )
    return matches[0]


# Daftar file yang dikenali otomatis.
DEM_FILE = find_file(
    DATA_DIR, ["dem", "elevasi", "elevation", "srtm"],
    {".tif", ".tiff"}
)

RAIN_FILE = find_file(
    DATA_DIR, ["hujan", "rainfall", "curah_hujan", "precip"],
    {".tif", ".tiff"}
)

BOUNDARY_FILE = find_file(
    DATA_DIR, ["batas", "boundary", "administrasi", "wilayah"],
    {".shp", ".gpkg", ".geojson"}
)

LITHOLOGY_FILE = find_file(
    DATA_DIR, ["litologi", "lithology", "geologi", "geology"],
    {".shp", ".gpkg", ".geojson"}
)

LULC_FILE = find_file(
    DATA_DIR, ["lulc", "landcover", "land_cover",
               "tutupan_lahan", "penggunaan_lahan"],
    {".shp", ".gpkg", ".geojson"}
)

RIVER_FILE = find_file(
    DATA_DIR, ["sungai", "river", "drainage", "hidrografi"],
    {".shp", ".gpkg", ".geojson"}
)

files = {
    "DEM": DEM_FILE,
    "Curah hujan": RAIN_FILE,
    "Batas wilayah": BOUNDARY_FILE,
    "Litologi": LITHOLOGY_FILE,
    "Tutupan lahan": LULC_FILE,
    "Sungai": RIVER_FILE,
}

print("\nDAFTAR FILE INPUT")
for name, file in files.items():
    print(f"{name:18s}: {file if file else 'BELUM DITEMUKAN'}")

missing = [name for name, file in files.items() if file is None]

if missing:
    raise FileNotFoundError(
        "\nFile berikut belum ditemukan: "
        + ", ".join(missing)
        + "\nPeriksa nama file di folder data atau sesuaikan "
          "kata kunci pada bagian 2 kode."
    )


# ================================================================
# 3. MEMBACA DEM DAN MENYIAPKAN GRID ANALISIS
# ================================================================

with rasterio.open(DEM_FILE) as src:
    dem_original = src.read(1, masked=True).astype("float32")
    dem_transform = src.transform
    dem_crs = src.crs
    dem_profile = src.profile.copy()
    dem_bounds = src.bounds
    dem_nodata = src.nodata

if dem_crs is None:
    raise ValueError(
        "DEM tidak memiliki CRS. Tetapkan CRS yang benar "
        "sebelum melanjutkan analisis."
    )

# DEM harus sudah berada pada CRS proyeksi dengan satuan meter.
# Kode tidak menebak CRS asli jika data belum terproyeksi.
if not dem_crs.is_projected:
    raise ValueError(
        f"CRS DEM adalah {dem_crs}, bukan CRS proyeksi. "
        "Proyeksikan DEM ke CRS metrik yang sesuai terlebih dahulu, "
        "kemudian sesuaikan TARGET_CRS jika diperlukan."
    )

if dem_crs.to_string() != TARGET_CRS:
    raise ValueError(
        f"CRS DEM ({dem_crs}) berbeda dari TARGET_CRS "
        f"({TARGET_CRS}). Proyeksikan DEM ke {TARGET_CRS}, "
        "atau ubah TARGET_CRS sesuai CRS proyeksi DEM."
    )

dem = dem_original.filled(np.nan)
valid_dem = np.isfinite(dem)

if dem_nodata is not None:
    valid_dem &= dem != dem_nodata

if not valid_dem.any():
    raise ValueError("Tidak ada piksel elevasi DEM yang valid.")

height, width = dem.shape
pixel_x = abs(dem_transform.a)
pixel_y = abs(dem_transform.e)

if pixel_x <= 0 or pixel_y <= 0:
    raise ValueError("Resolusi DEM tidak valid.")

print("\nInformasi DEM")
print("Ukuran raster :", dem.shape)
print("CRS           :", dem_crs)
print("Resolusi X    :", pixel_x, "meter")
print("Resolusi Y    :", pixel_y, "meter")


# ================================================================
# 4. MEMBACA DAN MENYIAPKAN DATA VEKTOR
# ================================================================

def read_vector(path, label):
    gdf = gpd.read_file(path)

    if gdf.empty:
        raise ValueError(f"Data {label} kosong.")

    if gdf.crs is None:
        raise ValueError(
            f"Data {label} tidak memiliki CRS. "
            "Tetapkan CRS sumber yang benar terlebih dahulu."
        )

    gdf = gdf.to_crs(TARGET_CRS)
    gdf = gdf[gdf.geometry.notna()].copy()
    gdf = gdf[~gdf.geometry.is_empty].copy()

    if gdf.empty:
        raise ValueError(f"Geometri data {label} tidak valid.")

    return gdf


boundary = read_vector(BOUNDARY_FILE, "batas wilayah")
lithology = read_vector(LITHOLOGY_FILE, "litologi")
lulc = read_vector(LULC_FILE, "tutupan lahan")
rivers = read_vector(RIVER_FILE, "sungai")

# Potong data ke area DEM agar pemrosesan lebih efisien.
dem_polygon = gpd.GeoDataFrame(
    {"id": [1]},
    geometry=[
        __import__("shapely.geometry", fromlist=["box"]).box(
            dem_bounds.left,
            dem_bounds.bottom,
            dem_bounds.right,
            dem_bounds.top
        )
    ],
    crs=TARGET_CRS
)

boundary = gpd.clip(boundary, dem_polygon)
lithology = gpd.clip(lithology, dem_polygon)
lulc = gpd.clip(lulc, dem_polygon)
rivers = gpd.clip(rivers, dem_polygon)

if boundary.empty or lithology.empty or lulc.empty or rivers.empty:
    raise ValueError(
        "Salah satu data vektor tidak beririsan dengan area DEM. "
        "Periksa cakupan spasial dan CRS seluruh data."
    )


# ================================================================
# 5. FUNGSI RASTERISASI
# ================================================================

def rasterize_vector(gdf, value_column, fill_value=np.nan):
    """
    Mengubah atribut numerik dari vektor menjadi raster
    dengan ukuran dan transformasi yang sama dengan DEM.
    """
    shapes = [
        (geom, float(value))
        for geom, value in zip(gdf.geometry, gdf[value_column])
        if geom is not None
        and not geom.is_empty
        and pd.notna(value)
    ]

    if not shapes:
        raise ValueError(
            f"Tidak ada geometri bernilai valid pada {value_column}."
        )

    return rasterize(
        shapes=shapes,
        out_shape=(height, width),
        transform=dem_transform,
        fill=fill_value,
        dtype="float32",
        all_touched=False
    )


# ================================================================
# 6. ANALISIS KEMIRINGAN LERENG
# ================================================================

# Menghitung gradien elevasi dalam arah X dan Y.
dem_for_gradient = np.where(valid_dem, dem, np.nan)

# Isi sementara piksel kosong dengan nilai valid terdekat
# agar perhitungan gradien tidak menyebarkan NaN.
from scipy.ndimage import distance_transform_edt

nearest_indices = distance_transform_edt(
    ~valid_dem, return_distances=False, return_indices=True
)

dem_filled = dem_for_gradient.copy()
dem_filled[~valid_dem] = dem_for_gradient[
    tuple(index[~valid_dem] for index in nearest_indices)
]

grad_y, grad_x = np.gradient(
    dem_filled, pixel_y, pixel_x
)

slope_deg = np.degrees(
    np.arctan(np.sqrt(grad_x**2 + grad_y**2))
)

# Klasifikasi skor lereng 1–5.
slope_score = np.select(
    [
        slope_deg < 8,
        slope_deg < 15,
        slope_deg < 25,
        slope_deg < 40
    ],
    [1, 2, 3, 4],
    default=5
).astype("float32")

slope_score[~valid_dem] = np.nan


# ================================================================
# 7. SKOR LITOLOGI DAN TUTUPAN LAHAN
# ================================================================

def find_attribute_column(gdf, candidates):
    """
    Memilih kolom atribut kategori berdasarkan nama kolom.
    Jika tidak ditemukan, gunakan kolom teks pertama yang tersedia.
    """
    for candidate in candidates:
        for column in gdf.columns:
            if candidate in column.lower():
                return column

    text_columns = [
        c for c in gdf.columns
        if c != gdf.geometry.name
        and (
            pd.api.types.is_object_dtype(gdf[c])
            or pd.api.types.is_string_dtype(gdf[c])
        )
    ]

    if text_columns:
        return text_columns[0]

    raise ValueError(
        "Kolom kategori tidak ditemukan. "
        "Periksa atribut tabel data vektor."
    )


lith_col = find_attribute_column(
    lithology,
    ["litologi", "lithology", "batuan", "jenis", "geologi", "formation"]
)

lulc_col = find_attribute_column(
    lulc,
    ["lulc", "landcover", "tutupan", "penggunaan", "kelas", "jenis"]
)

print("\nKolom kategori yang digunakan")
print("Litologi      :", lith_col)
print("Tutupan lahan :", lulc_col)


def score_lithology(value):
    """
    Skor indikatif berdasarkan nama batuan.
    Sesuaikan dengan karakteristik geologi setempat.
    """
    s = str(value).lower()

    if any(k in s for k in [
        "alluv", "aluvial", "lempung", "clay",
        "colluv", "koluv"
    ]):
        return 5

    if any(k in s for k in [
        "lapuk", "weathered", "breksi", "breccia",
        "tuf", "tuff", "serpih", "shale"
    ]):
        return 4

    if any(k in s for k in [
        "sedimen", "sedimentary", "batupasir",
        "sandstone", "konglomerat", "conglomerate"
    ]):
        return 3

    if any(k in s for k in [
        "andesit", "andesite", "basalt", "basal",
        "vulkanik", "volcanic"
    ]):
        return 2

    if any(k in s for k in [
        "granit", "granite", "diorit", "diorite",
        "metamorf", "metamorphic", "gneiss", "gabbro"
    ]):
        return 1

    return 3


def score_lulc(value):
    """
    Skor indikatif berdasarkan kelas tutupan/penggunaan lahan.
    Nilai lebih tinggi berarti kondisi relatif lebih mendukung
    kerentanan dalam model awal, bukan probabilitas longsor.
    """
    s = str(value).lower()

    if any(k in s for k in [
        "bare", "tanah terbuka", "lahan terbuka",
        "open land", "tambang", "mining"
    ]):
        return 5

    if any(k in s for k in [
        "pertanian", "agriculture", "tegalan",
        "cropland", "perkebunan", "plantation"
    ]):
        return 4

    if any(k in s for k in [
        "semak", "shrub", "grass", "rumput",
        "sawah", "paddy"
    ]):
        return 3

    if any(k in s for k in [
        "permukiman", "permukiman", "built-up",
        "urban", "settlement"
    ]):
        return 3

    if any(k in s for k in [
        "hutan", "forest", "mangrove", "vegetasi rapat"
    ]):
        return 1

    if any(k in s for k in [
        "air", "water", "danau", "lake"
    ]):
        return 1

    return 3


lithology["_score"] = lithology[lith_col].apply(score_lithology)
lulc["_score"] = lulc[lulc_col].apply(score_lulc)

lith_score = rasterize_vector(lithology, "_score")
lulc_score = rasterize_vector(lulc, "_score")

lith_score[lith_score == 0] = np.nan
lulc_score[lulc_score == 0] = np.nan


# ================================================================
# 8. MEMPROSES DATA CURAH HUJAN
# ================================================================

with rasterio.open(RAIN_FILE) as src:
    rain_source = src.read(1, masked=True).astype("float32")
    rain_crs = src.crs
    rain_transform = src.transform
    rain_nodata = src.nodata

if rain_crs is None:
    raise ValueError("Raster curah hujan tidak memiliki CRS.")

rain_source_data = rain_source.filled(np.nan)

if rain_nodata is not None:
    rain_source_data[
        rain_source_data == rain_nodata
    ] = np.nan

rain_aligned = np.full(
    (height, width), np.nan, dtype="float32"
)

reproject(
    source=rain_source_data,
    destination=rain_aligned,
    src_transform=rain_transform,
    src_crs=rain_crs,
    src_nodata=np.nan,
    dst_transform=dem_transform,
    dst_crs=dem_crs,
    dst_nodata=np.nan,
    resampling=Resampling.bilinear
)

rain_valid = (
    valid_dem
    & np.isfinite(rain_aligned)
    & (rain_aligned >= 0)
)

if not rain_valid.any():
    raise ValueError(
        "Data curah hujan tidak memiliki piksel valid "
        "pada area DEM."
    )

# Klasifikasi curah hujan menggunakan kuantil data setempat.
# Ini klasifikasi relatif, bukan ambang hujan universal.
rain_values = rain_aligned[rain_valid]

q20, q40, q60, q80 = np.quantile(
    rain_values, [0.20, 0.40, 0.60, 0.80]
)

rain_score = np.full(
    (height, width), np.nan, dtype="float32"
)

rain_score[rain_valid] = np.digitize(
    rain_aligned[rain_valid],
    bins=[q20, q40, q60, q80]
) + 1


# ================================================================
# 9. ANALISIS JARAK TERHADAP SUNGAI
# ================================================================

river_shapes = [
    (geom, 1)
    for geom in rivers.geometry
    if geom is not None and not geom.is_empty
]

river_mask = rasterize(
    river_shapes,
    out_shape=(height, width),
    transform=dem_transform,
    fill=0,
    dtype="uint8",
    all_touched=True
)

if river_mask.sum() == 0:
    raise ValueError(
        "Sungai tidak terwakili dalam grid DEM. "
        "Periksa posisi sungai dan cakupan DEM."
    )

# Menghitung jarak setiap piksel ke piksel sungai terdekat.
# Jarak dihitung dalam meter.
distance_to_river = distance_transform_edt(
    river_mask == 0,
    sampling=(pixel_y, pixel_x)
)

# Skor indikatif: semakin dekat ke sungai, skor semakin tinggi.
river_score = np.select(
    [
        distance_to_river <= 50,
        distance_to_river <= 100,
        distance_to_river <= 250,
        distance_to_river <= 500
    ],
    [5, 4, 3, 2],
    default=1
).astype("float32")

river_score[~valid_dem] = np.nan


# ================================================================
# 10. WEIGHTED OVERLAY / LANDSLIDE SUSCEPTIBILITY INDEX
# ================================================================

# Setiap parameter memiliki skor 1–5.
# Bobot total = 1.
parameter_scores = {
    "slope": slope_score,
    "lithology": lith_score,
    "lulc": lulc_score,
    "rainfall": rain_score,
    "river_distance": river_score
}

# Piksel valid harus memiliki seluruh parameter.
valid_all = valid_dem.copy()

for score in parameter_scores.values():
    valid_all &= np.isfinite(score)

if not valid_all.any():
    raise ValueError(
        "Tidak ada piksel yang memiliki seluruh parameter valid. "
        "Periksa tumpang tindih dan cakupan data."
    )

lsi = np.full((height, width), np.nan, dtype="float32")

weighted_sum = np.zeros((height, width), dtype="float32")

for parameter, score in parameter_scores.items():
    weighted_sum[valid_all] += (
        WEIGHTS[parameter] * score[valid_all]
    )

lsi[valid_all] = weighted_sum[valid_all]

# Rentang teoritis LSI: 1–5.
# Klasifikasi awal berdasarkan rentang indeks.
class_edges = [1, 1.8, 2.6, 3.4, 4.2, 5.01]

class_names = [
    "Sangat Rendah",
    "Rendah",
    "Sedang",
    "Tinggi",
    "Sangat Tinggi"
]

susceptibility = np.full(
    (height, width), 0, dtype="uint8"
)

susceptibility[valid_all] = np.digitize(
    lsi[valid_all],
    bins=class_edges[1:-1]
) + 1

# Nilai 0 digunakan untuk piksel tanpa data.
print("\nPembobotan parameter:")
for parameter, weight in WEIGHTS.items():
    print(f"{parameter:18s}: {weight:.0%}")

print("Piksel valid:", int(valid_all.sum()))


# ================================================================
# 11. STATISTIK LUAS SETIAP KELAS
# ================================================================

pixel_area_m2 = pixel_x * pixel_y
statistics = []

for class_id, class_name in enumerate(class_names, start=1):
    count = int(np.sum(susceptibility == class_id))
    area_m2 = count * pixel_area_m2

    statistics.append({
        "Kelas": class_name,
        "Kode": class_id,
        "Jumlah_Piksel": count,
        "Luas_m2": area_m2,
        "Luas_ha": area_m2 / 10000,
        "Luas_km2": area_m2 / 1000000,
        "Persentase": (
            count / int(valid_all.sum()) * 100
            if valid_all.sum() else 0
        )
    })

stats_df = pd.DataFrame(statistics)

print("\nSTATISTIK KERENTANAN LONGSOR")
print(stats_df.round(3).to_string(index=False))

stats_df.to_csv(
    OUT_DIR / "statistik_luas_kerentanan.csv",
    index=False
)


# ================================================================
# 12. EKSPOR RASTER GEOTIFF
# ================================================================

output_profile = dem_profile.copy()
output_profile.update(
    driver="GTiff",
    height=height,
    width=width,
    count=1,
    dtype="float32",
    crs=dem_crs,
    transform=dem_transform,
    nodata=-9999.0,
    compress="lzw"
)


def save_float_raster(array, filename):
    output = np.where(
        np.isfinite(array), array, -9999.0
    ).astype("float32")

    with rasterio.open(
        OUT_DIR / filename,
        "w",
        **output_profile
    ) as dst:
        dst.write(output, 1)


save_float_raster(
    slope_deg, "kemiringan_lereng_derajat.tif"
)

save_float_raster(
    lsi, "indeks_kerentanan_longsor.tif"
)

save_float_raster(
    distance_to_river.astype("float32"),
    "jarak_ke_sungai_meter.tif"
)

# Simpan skor masing-masing parameter untuk audit.
for name, score in parameter_scores.items():
    save_float_raster(
        score, f"skor_{name}.tif"
    )

class_profile = output_profile.copy()
class_profile.update(
    dtype="uint8",
    nodata=0
)

with rasterio.open(
    OUT_DIR / "kelas_kerentanan_longsor.tif",
    "w",
    **class_profile
) as dst:
    dst.write(susceptibility, 1)


# ================================================================
# 13. VISUALISASI PETA
# ================================================================

colors = [
    "#2CA25F",  # Sangat rendah
    "#A1D99B",  # Rendah
    "#FFFF99",  # Sedang
    "#FD8D3C",  # Tinggi
    "#BD0026"   # Sangat tinggi
]

cmap = ListedColormap(colors)
cmap.set_bad("white")

class_norm = BoundaryNorm(
    np.arange(0.5, 6.5, 1),
    cmap.N
)

fig, axes = plt.subplots(
    2, 2, figsize=(16, 13), constrained_layout=True
)

# DEM
im0 = axes[0, 0].imshow(
    np.ma.masked_invalid(dem),
    cmap="terrain"
)
axes[0, 0].set_title("Elevasi DEM (m)")
plt.colorbar(im0, ax=axes[0, 0], shrink=0.8, label="Elevasi (m)")

# Kemiringan
im1 = axes[0, 1].imshow(
    np.ma.masked_invalid(slope_deg),
    cmap="YlOrRd",
    vmin=0
)
axes[0, 1].set_title("Kemiringan Lereng (derajat)")
plt.colorbar(
    im1, ax=axes[0, 1], shrink=0.8, label="Derajat"
)

# Jarak sungai
im2 = axes[1, 0].imshow(
    np.ma.masked_invalid(distance_to_river),
    cmap="Blues"
)
axes[1, 0].set_title("Jarak ke Sungai (m)")
plt.colorbar(
    im2, ax=axes[1, 0], shrink=0.8, label="Meter"
)

# Peta kerentanan
masked_classes = np.ma.masked_where(
    susceptibility == 0, susceptibility
)

im3 = axes[1, 1].imshow(
    masked_classes,
    cmap=cmap,
    norm=class_norm,
    interpolation="nearest"
)
axes[1, 1].set_title("Peta Kerentanan Longsor")

cbar = plt.colorbar(
    im3,
    ax=axes[1, 1],
    ticks=[1, 2, 3, 4, 5],
    shrink=0.8
)
cbar.ax.set_yticklabels(class_names)

for ax in axes.flat:
    ax.set_xlabel("Kolom piksel")
    ax.set_ylabel("Baris piksel")

fig.suptitle(
    "PEMETAAN KERENTANAN LONGSOR BERBASIS GIS",
    fontsize=16,
    fontweight="bold"
)

fig.savefig(
    OUT_DIR / "peta_parameter_dan_kerentanan.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ================================================================
# 14. PETA KERENTANAN LONGSOR DENGAN BATAS WILAYAH
# ================================================================

fig, ax = plt.subplots(figsize=(12, 10))

ax.imshow(
    masked_classes,
    extent=[
        dem_bounds.left,
        dem_bounds.right,
        dem_bounds.bottom,
        dem_bounds.top
    ],
    origin="upper",
    cmap=cmap,
    norm=class_norm,
    interpolation="nearest"
)

boundary.boundary.plot(
    ax=ax,
    color="black",
    linewidth=1.0
)

rivers.plot(
    ax=ax,
    color="blue",
    linewidth=0.5,
    alpha=0.7
)

legend_handles = [
    plt.Line2D(
        [0], [0],
        color=colors[i],
        linewidth=8,
        label=class_names[i]
    )
    for i in range(5)
]

ax.legend(
    handles=legend_handles,
    title="Kelas Kerentanan",
    loc="lower left"
)

ax.set_title(
    "Peta Kerentanan Longsor Berbasis GIS",
    fontsize=15,
    fontweight="bold"
)
ax.set_xlabel("Easting (m)")
ax.set_ylabel("Northing (m)")
ax.set_aspect("equal")

plt.tight_layout()
plt.savefig(
    OUT_DIR / "peta_kerentanan_longsor.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()


# ================================================================
# 15. EKSPOR KONFIGURASI DAN METADATA
# ================================================================

metadata = {
    "judul": "Pemetaan Kerentanan Longsor Berbasis GIS",
    "folder_data": str(DATA_DIR),
    "folder_output": str(OUT_DIR),
    "crs": str(dem_crs),
    "ukuran_raster": [int(height), int(width)],
    "resolusi_meter": [pixel_x, pixel_y],
    "bobot_parameter": WEIGHTS,
    "klasifikasi_lsi": {
        "rentang": class_edges,
        "kelas": class_names
    },
    "ambang_kuantil_curah_hujan": {
        "q20": float(q20),
        "q40": float(q40),
        "q60": float(q60),
        "q80": float(q80)
    },
    "file_input": {
        key: str(value) for key, value in files.items()
    },
    "catatan": (
        "Hasil merupakan indeks kerentanan relatif berbasis "
        "weighted overlay, bukan probabilitas longsor. "
        "Bobot dan skor harus divalidasi dengan data lokal."
    )
}

with open(
    OUT_DIR / "metadata_analisis.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(metadata, f, indent=4, ensure_ascii=False)


# ================================================================
# 16. RINGKASAN HASIL
# ================================================================

print("\n" + "=" * 65)
print("ANALISIS SELESAI")
print("=" * 65)
print("Lokasi output:", OUT_DIR)

print("\nFile hasil yang dibuat:")
for output_file in sorted(OUT_DIR.iterdir()):
    if output_file.is_file():
        print("-", output_file.name)

print(
    "\nCatatan metodologi:\n"
    "- Kelas lereng menggunakan ambang derajat yang ditetapkan.\n"
    "- Skor litologi dan tutupan lahan bersifat indikatif.\n"
    "- Curah hujan dikelaskan berdasarkan kuantil data.\n"
    "- Skor jarak sungai merupakan asumsi awal untuk pemodelan.\n"
    "- Bobot dan klasifikasi perlu diuji dengan data longsor "
    "historis atau validasi lapangan.\n"
)
````

````python

# ================================================================
# WEB GIS UNTUK INTEGRASI DATA INVESTIGASI GEOTEKNIK
# Python + GeoPandas + Folium
# Output: peta HTML interaktif, GeoJSON, laporan QC
# ================================================================

from pathlib import Path
import json
import html
import re
import warnings

import numpy as np
import pandas as pd
import geopandas as gpd
import folium

from folium.plugins import Search, Fullscreen, MeasureControl
from branca.element import Element

warnings.filterwarnings("ignore", category=UserWarning)

# ================================================================
# 1. KONFIGURASI
# ================================================================

BASE_DIR = Path(
    r"D:\KULIAH\SEMESTER 7\KB PART 2\sample_data_GIS_Longsor"
)

DATA_DIR = BASE_DIR / "data"
OUT_DIR = BASE_DIR / "output_webgis"
OUT_DIR.mkdir(parents=True, exist_ok=True)

# CRS tujuan peta dan ekspor GeoJSON.
WEB_CRS = "EPSG:4326"

# Isi dengan CRS sumber tabel X-Y jika koordinatnya berupa
# Easting/Northing, bukan Longitude/Latitude.
# Contoh: "EPSG:32750" HANYA jika CRS sumber memang UTM 50S.
# Biarkan None jika CRS tabel X-Y belum diketahui.
TABLE_XY_CRS = None

# Jika nama kolom tidak dikenali otomatis, isi nama sebenarnya.
# Contoh: ID_COLUMN = "ID_Titik"
ID_COLUMN = None
LAT_COLUMN = None
LON_COLUMN = None
X_COLUMN = None
Y_COLUMN = None

# CRS sumber untuk file vektor harus tercantum dengan benar
# di metadata file. Jangan menebak CRS jika tidak tersedia.

print("=" * 65)
print("WEB GIS INVESTIGASI GEOTEKNIK")
print("=" * 65)
print("Data   :", DATA_DIR)
print("Output :", OUT_DIR)

if not DATA_DIR.exists():
    raise FileNotFoundError(
        f"Folder data tidak ditemukan: {DATA_DIR}"
    )


# ================================================================
# 2. UTILITAS PEMBERSIHAN NAMA KOLOM
# ================================================================

def normalize_name(value):
    return re.sub(
        r"[^a-z0-9]+", "_", str(value).strip().lower()
    ).strip("_")


def find_column(df, candidates, override=None):
    """
    Cari kolom berdasarkan nama alternatif.
    Override digunakan jika nama kolom diketahui secara pasti.
    """
    if override:
        if override in df.columns:
            return override
        raise KeyError(
            f"Kolom konfigurasi '{override}' tidak ditemukan. "
            f"Kolom tersedia: {list(df.columns)}"
        )

    normalized = {
        normalize_name(col): col for col in df.columns
    }

    for candidate in candidates:
        key = normalize_name(candidate)
        if key in normalized:
            return normalized[key]

    # Mencari kecocokan sebagian nama kolom.
    for candidate in candidates:
        key = normalize_name(candidate)
        for normalized_col, original_col in normalized.items():
            if key and key in normalized_col:
                return original_col

    return None


def detect_id_column(df):
    return find_column(
        df,
        [
            "id_titik", "id", "kode_titik", "nama_titik",
            "id_investigasi", "borehole_id", "hole_id",
            "point_id", "station", "nama"
        ],
        ID_COLUMN
    )


def detect_coordinates(df):
    lat_col = find_column(
        df,
        ["latitude", "lat", "lintang", "y_lat"],
        LAT_COLUMN
    )
    lon_col = find_column(
        df,
        ["longitude", "lon", "long", "bujur", "x_lon"],
        LON_COLUMN
    )

    if lat_col and lon_col:
        return "geographic", lat_col, lon_col

    x_col = find_column(
        df,
        ["easting", "utm_easting", "x_coord", "x_coordinate", "x"],
        X_COLUMN
    )
    y_col = find_column(
        df,
        ["northing", "utm_northing", "y_coord", "y_coordinate", "y"],
        Y_COLUMN
    )

    if x_col and y_col:
        return "projected", x_col, y_col

    return None, None, None


def classify_investigation(filename, df):
    """
    Mengidentifikasi jenis investigasi dari nama file
    dan, jika perlu, nama kolom.
    """
    name = normalize_name(filename)

    if any(k in name for k in ["borehole", "bore_hole", "bh_",
                               "bor", "drilling"]):
        return "Borehole"

    if any(k in name for k in ["cpt", "sondir", "cone_penetration"]):
        return "CPT / Sondir"

    if any(k in name for k in ["masw", "surface_wave"]):
        return "MASW"

    if any(k in name for k in ["ert", "resistivity", "resistivitas"]):
        return "ERT"

    if any(k in name for k in ["hvsr", "microtremor"]):
        return "HVSR"

    if any(k in name for k in ["spt", "standard_penetration"]):
        return "SPT"

    cols = " ".join(normalize_name(c) for c in df.columns)

    if "spt_n" in cols or "n_spt" in cols:
        return "SPT"

    if "resistivitas" in cols or "resistivity" in cols:
        return "ERT"

    if "vs30" in cols:
        return "MASW / Vs30"

    return "Titik investigasi"


# ================================================================
# 3. MEMBACA CSV DAN EXCEL
# ================================================================

tabular_files = [
    p for p in DATA_DIR.rglob("*")
    if p.is_file()
    and p.suffix.lower() in {".csv", ".xlsx", ".xls"}
    and OUT_DIR not in p.parents
]

point_layers = {}
qc_records = []
source_records = []

# CRS dan koordinat tabel diperiksa sebelum dibuat menjadi titik.
for file in tabular_files:
    try:
        if file.suffix.lower() == ".csv":
            df = pd.read_csv(file)
        else:
            df = pd.read_excel(file)

        if df.empty:
            qc_records.append({
                "file": file.name,
                "status": "GAGAL",
                "masalah": "Tabel kosong"
            })
            continue

        df.columns = [str(c).strip() for c in df.columns]

        id_col = detect_id_column(df)
        coord_type, coord_1, coord_2 = detect_coordinates(df)

        if coord_type is None:
            qc_records.append({
                "file": file.name,
                "status": "GAGAL",
                "masalah": "Kolom koordinat tidak ditemukan"
            })
            continue

        # ID unik: jangan diam-diam menggabungkan titik berbeda.
        if id_col:
            duplicate_ids = int(
                df[id_col].astype("string").duplicated(
                    keep=False
                ).sum()
            )
            missing_ids = int(df[id_col].isna().sum())
        else:
            duplicate_ids = None
            missing_ids = len(df)

        x = pd.to_numeric(df[coord_1], errors="coerce")
        y = pd.to_numeric(df[coord_2], errors="coerce")

        valid = x.notna() & y.notna()

        if coord_type == "geographic":
            valid &= x.between(-90, 90) & y.between(-180, 180)

        if not valid.any():
            qc_records.append({
                "file": file.name,
                "status": "GAGAL",
                "masalah": "Tidak ada koordinat valid"
            })
            continue

        cleaned = df.loc[valid].copy()
        cleaned[coord_1] = x.loc[valid]
        cleaned[coord_2] = y.loc[valid]

        if coord_type == "geographic":
            gdf = gpd.GeoDataFrame(
                cleaned,
                geometry=gpd.points_from_xy(
                    cleaned[coord_2],
                    cleaned[coord_1]
                ),
                crs="EPSG:4326"
            )
        else:
            if TABLE_XY_CRS is None:
                qc_records.append({
                    "file": file.name,
                    "status": "PERLU KONFIGURASI",
                    "masalah": (
                        "Koordinat X-Y terdeteksi, tetapi "
                        "TABLE_XY_CRS belum ditentukan"
                    )
                })
                continue

            gdf = gpd.GeoDataFrame(
                cleaned,
                geometry=gpd.points_from_xy(
                    cleaned[coord_1],
                    cleaned[coord_2]
                ),
                crs=TABLE_XY_CRS
            )

            if not gdf.crs.is_projected:
                raise ValueError(
                    "TABLE_XY_CRS harus CRS proyeksi untuk X-Y."
                )

            gdf = gdf.to_crs(WEB_CRS)

        if id_col is None:
            gdf["ID_WebGIS"] = [
                f"{file.stem}_{i+1:04d}"
                for i in range(len(gdf))
            ]
            id_col = "ID_WebGIS"

        # Pertahankan ID asli; ID duplikat dilaporkan.
        # Marker dengan ID sama akan memiliki ID internal unik.
        gdf["_source_id"] = gdf[id_col].astype("string")
        gdf["_webgis_id"] = [
            f"{file.stem}_{i+1:05d}"
            for i in range(len(gdf))
        ]
        gdf["_jenis_investigasi"] = classify_investigation(
            file.stem, gdf
        )
        gdf["_sumber_file"] = str(file.name)

        layer_name = f"{file.stem} ({len(gdf)} titik)"
        point_layers[layer_name] = gdf

        qc_records.append({
            "file": file.name,
            "status": "OK",
            "jumlah_baris": len(df),
            "koordinat_valid": int(valid.sum()),
            "koordinat_ditolak": int((~valid).sum()),
            "id_duplikat_baris": duplicate_ids,
            "id_kosong": missing_ids,
            "kolom_id": id_col,
            "kolom_koordinat_1": coord_1,
            "kolom_koordinat_2": coord_2,
            "jenis_investigasi": gdf[
                "_jenis_investigasi"
            ].iloc[0]
        })

        source_records.append({
            "file": file.name,
            "jenis": "Tabel",
            "jumlah_fitur": len(gdf)
        })

        print(f"[OK] {file.name}: {len(gdf)} titik")

    except Exception as exc:
        qc_records.append({
            "file": file.name,
            "status": "GAGAL",
            "masalah": str(exc)
        })
        print(f"[GAGAL] {file.name}: {exc}")


# ================================================================
# 4. MEMBACA SHAPEFILE, GEOPACKAGE, DAN GEOJSON
# ================================================================

vector_files = [
    p for p in DATA_DIR.rglob("*")
    if p.is_file()
    and p.suffix.lower() in {".shp", ".gpkg", ".geojson"}
    and OUT_DIR not in p.parents
]

vector_layers = {}

for file in vector_files:
    try:
        gdf = gpd.read_file(file)

        if gdf.empty:
            qc_records.append({
                "file": file.name,
                "status": "GAGAL",
                "masalah": "Layer vektor kosong"
            })
            continue

        if gdf.crs is None:
            qc_records.append({
                "file": file.name,
                "status": "GAGAL",
                "masalah": "CRS sumber tidak tersedia"
            })
            continue

        original_crs = str(gdf.crs)
        gdf = gdf.to_crs(WEB_CRS)

        geom_types = set(
            gdf.geometry.geom_type.dropna().tolist()
        )

        # Titik dari SHP juga dapat menjadi layer investigasi.
        if geom_types and geom_types.issubset(
            {"Point", "MultiPoint"}
        ):
            gdf["_sumber_file"] = file.name
            gdf["_jenis_investigasi"] = classify_investigation(
                file.stem, gdf
            )
            gdf["_webgis_id"] = [
                f"{file.stem}_{i+1:05d}"
                for i in range(len(gdf))
            ]

            id_col = detect_id_column(gdf)
            if id_col:
                gdf["_source_id"] = gdf[id_col].astype("string")
            else:
                gdf["_source_id"] = gdf["_webgis_id"]

            layer_name = f"{file.stem} ({len(gdf)} titik)"
            point_layers[layer_name] = gdf

        else:
            # Layer poligon dan garis disimpan terpisah.
            vector_layers[file.stem] = gdf

        qc_records.append({
            "file": file.name,
            "status": "OK",
            "jumlah_fitur": len(gdf),
            "crs_sumber": original_crs,
            "crs_web": WEB_CRS,
            "jenis_geometri": ", ".join(sorted(geom_types))
        })

        source_records.append({
            "file": file.name,
            "jenis": "Vektor",
            "jumlah_fitur": len(gdf)
        })

        print(f"[OK] Layer vektor: {file.name}")

    except Exception as exc:
        qc_records.append({
            "file": file.name,
            "status": "GAGAL",
            "masalah": str(exc)
        })
        print(f"[GAGAL] {file.name}: {exc}")


# ================================================================
# 5. MENYIMPAN HASIL QUALITY CONTROL
# ================================================================

qc_df = pd.DataFrame(qc_records)

qc_path = OUT_DIR / "laporan_quality_control.csv"
qc_df.to_csv(qc_path, index=False, encoding="utf-8-sig")

source_df = pd.DataFrame(source_records)
source_df.to_csv(
    OUT_DIR / "daftar_sumber_data.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nRINGKASAN QUALITY CONTROL")
if not qc_df.empty:
    print(qc_df.to_string(index=False))
else:
    print("Tidak ada file input yang dapat diproses.")

print("\nJumlah layer titik:", len(point_layers))
print("Jumlah layer vektor:", len(vector_layers))


# ================================================================
# 6. MENYIAPKAN PETA DASAR
# ================================================================

all_points = []

for layer_name, gdf in point_layers.items():
    if not gdf.empty:
        all_points.append(gdf.to_crs(WEB_CRS))

if all_points:
    combined_points = gpd.GeoDataFrame(
        pd.concat(all_points, ignore_index=True),
        geometry="geometry",
        crs=WEB_CRS
    )

    # Pusat peta dihitung dari data aktual.
    center_lat = combined_points.geometry.y.median()
    center_lon = combined_points.geometry.x.median()

elif vector_layers:
    first_layer = next(iter(vector_layers.values()))
    minx, miny, maxx, maxy = first_layer.total_bounds
    center_lat = (miny + maxy) / 2
    center_lon = (minx + maxx) / 2

else:
    raise ValueError(
        "Tidak ada data titik atau layer vektor yang berhasil "
        "dibaca. Periksa file input dan laporan QC."
    )

m = folium.Map(
    location=[center_lat, center_lon],
    zoom_start=13,
    tiles=None,
    control_scale=True
)

folium.TileLayer(
    tiles="OpenStreetMap",
    name="OpenStreetMap",
    control=True
).add_to(m)

folium.TileLayer(
    tiles="CartoDB positron",
    name="CartoDB Positron",
    control=True
).add_to(m)

Fullscreen(
    position="topleft",
    title="Layar penuh",
    title_cancel="Keluar layar penuh"
).add_to(m)

MeasureControl(
    position="topleft",
    primary_length_unit="meters",
    secondary_length_unit="kilometers"
).add_to(m)


# ================================================================
# 7. SIMBOL BERDASARKAN JENIS INVESTIGASI
# ================================================================

STYLE_BY_TYPE = {
    "Borehole": {"color": "red", "icon": "tint"},
    "SPT": {"color": "orange", "icon": "tasks"},
    "CPT / Sondir": {"color": "purple", "icon": "signal"},
    "MASW": {"color": "blue", "icon": "wave-square"},
    "MASW / Vs30": {"color": "blue", "icon": "wave-square"},
    "ERT": {"color": "green", "icon": "flash"},
    "HVSR": {"color": "cadetblue", "icon": "globe"},
    "Titik investigasi": {"color": "darkblue", "icon": "info-sign"}
}


def safe_value(value):
    """Mencegah teks atribut diinterpretasikan sebagai HTML."""
    if pd.isna(value):
        return "-"
    return html.escape(str(value))


def make_popup(row):
    """Membuat popup dari atribut asli, bukan nilai buatan."""
    excluded = {"geometry"}

    rows = []

    preferred = [
        "ID_Titik", "ID", "Nama_Titik", "Latitude", "Longitude",
        "Elevasi", "Kedalaman", "SPT_N", "Jenis_Tanah",
        "Vs30", "Resistivitas", "Muka_Air_Tanah", "Keterangan"
    ]

    keys = list(row.index)
    ordered_keys = []

    for wanted in preferred:
        for key in keys:
            if normalize_name(key) == normalize_name(wanted):
                if key not in ordered_keys:
                    ordered_keys.append(key)

    for key in keys:
        if key not in ordered_keys:
            ordered_keys.append(key)

    for key in ordered_keys:
        if key in excluded or str(key).startswith("_"):
            continue

        value = row[key]

        # Popup hanya berisi nilai atribut yang tersedia.
        if value is None or (
            isinstance(value, (float, np.floating))
            and not np.isfinite(value)
        ):
            continue

        rows.append(
            "<tr>"
            f"<th style='text-align:left;padding:4px;'>{safe_value(key)}</th>"
            f"<td style='padding:4px;'>{safe_value(value)}</td>"
            "</tr>"
        )

    lat = row.geometry.y
    lon = row.geometry.x

    content = f"""
    <div style="width:290px;max-height:350px;overflow:auto;">
      <h4 style="margin:4px 0 8px 0;">
        {safe_value(row.get('_source_id', row.get('_webgis_id', 'Titik')))}
      </h4>
      <p style="font-size:11px;">
        WGS 84 | Lat: {lat:.6f}, Lon: {lon:.6f}
      </p>
      <table style="width:100%;font-size:12px;border-collapse:collapse;">
        {''.join(rows)}
      </table>
    </div>
    """

    return folium.Popup(content, max_width=340)


# ================================================================
# 8. MENAMBAHKAN LAYER TITIK
# ================================================================

search_group = folium.FeatureGroup(
    name="Pencarian titik investigasi",
    show=True
)
search_features = []

for layer_name, gdf in point_layers.items():
    gdf = gdf.to_crs(WEB_CRS)

    if gdf.empty:
        continue

    fg = folium.FeatureGroup(name=layer_name, show=True)

    for _, row in gdf.iterrows():
        if row.geometry is None or row.geometry.is_empty:
            continue

        if row.geometry.geom_type != "Point":
            continue

        kind = str(
            row.get("_jenis_investigasi", "Titik investigasi")
        )
        style = STYLE_BY_TYPE.get(
            kind, STYLE_BY_TYPE["Titik investigasi"]
        )

        marker = folium.Marker(
            location=[
                row.geometry.y,
                row.geometry.x
            ],
            tooltip=(
                f"{kind}: "
                f"{row.get('_source_id', row.get('_webgis_id', 'Titik'))}"
            ),
            popup=make_popup(row),
            icon=folium.Icon(
                color=style["color"],
                icon=style["icon"],
                prefix="glyphicon"
            )
        )

        marker.add_to(fg)

        # Data pencarian berasal dari atribut aktual.
        props = {
            "ID": str(
                row.get("_source_id", row.get("_webgis_id", ""))
            ),
            "Jenis": kind,
            "Sumber": str(row.get("_sumber_file", ""))
        }

        search_features.append(
            {
                "type": "Feature",
                "geometry": {
                    "type": "Point",
                    "coordinates": [
                        float(row.geometry.x),
                        float(row.geometry.y)
                    ]
                },
                "properties": props
            }
        )

    fg.add_to(m)

# Layer khusus pencarian; fitur ini hanya digunakan oleh kontrol Search.
search_geojson = {
    "type": "FeatureCollection",
    "features": search_features
}

if search_features:
    folium.GeoJson(
        search_geojson,
        name="Indeks pencarian",
        show=False
    ).add_to(search_group)

    search_group.add_to(m)

    Search(
        layer=search_group,
        search_label="ID",
        placeholder="Cari ID titik...",
        collapsed=False,
        search_zoom=17
    ).add_to(m)


# ================================================================
# 9. MENAMBAHKAN LAYER PENDUKUNG
# ================================================================

def vector_style(feature):
    geom_type = feature["geometry"]["type"]

    if geom_type in ["Polygon", "MultiPolygon"]:
        return {
            "color": "#555555",
            "weight": 1.5,
            "fillColor": "#BDBDBD",
            "fillOpacity": 0.12
        }

    return {
        "color": "#555555",
        "weight": 2
    }


for name, gdf in vector_layers.items():
    if gdf.empty:
        continue

    fg = folium.FeatureGroup(
        name=f"Pendukung: {name}",
        show=False
    )

    # Simpan properti non-geometri agar popup menampilkan
    # atribut sumber.
    geojson = json.loads(
        gdf.to_json(drop_id=True)
    )

    def make_feature_popup(feature):
        props = feature.get("properties") or {}

        rows = []
        for key, value in props.items():
            if value is None:
                continue
            rows.append(
                "<tr>"
                f"<th style='text-align:left;padding:3px;'>{safe_value(key)}</th>"
                f"<td style='padding:3px;'>{safe_value(value)}</td>"
                "</tr>"
            )

        return (
            "<div style='max-height:300px;overflow:auto;'>"
            f"<h4>{safe_value(name)}</h4>"
            "<table>"
            + "".join(rows)
            + "</table></div>"
        )

    folium.GeoJson(
        geojson,
        name=name,
        style_function=vector_style,
        tooltip=folium.GeoJsonTooltip(
            fields=[
                c for c in gdf.columns
                if c != gdf.geometry.name
            ][:3]
        ) if any(
            c != gdf.geometry.name for c in gdf.columns
        ) else None,
        popup=folium.GeoJsonPopup(
            fields=[
                c for c in gdf.columns
                if c != gdf.geometry.name
            ][:12],
            labels=True
        ) if any(
            c != gdf.geometry.name for c in gdf.columns
        ) else None
    ).add_to(fg)

    fg.add_to(m)


# ================================================================
# 10. LEGENDA DAN INFORMASI PETA
# ================================================================

legend_html = """
<div style="
position: fixed;
bottom: 30px;
left: 15px;
z-index: 9999;
background: white;
border: 1px solid #888;
border-radius: 5px;
padding: 10px 12px;
font-size: 12px;
box-shadow: 0 1px 5px rgba(0,0,0,0.25);
">
<b>LEGENDA INVESTIGASI</b><br>
<span style="color:red;">●</span> Borehole<br>
<span style="color:orange;">●</span> SPT<br>
<span style="color:purple;">●</span> CPT / Sondir<br>
<span style="color:blue;">●</span> MASW / Vs30<br>
<span style="color:green;">●</span> ERT<br>
<span style="color:cadetblue;">●</span> HVSR<br>
<hr style="margin:5px 0;">
<b>CRS tampilan:</b> WGS 84 (EPSG:4326)<br>
<b>Sumber:</b> File input pada folder data
</div>
"""

m.get_root().html.add_child(Element(legend_html))

# Layer control harus ditambahkan setelah seluruh layer.
folium.LayerControl(
    collapsed=False,
    position="topright"
).add_to(m)


# ================================================================
# 11. FITUR FILTER BERDASARKAN ATRIBUT
# ================================================================

# Filter ini memfilter marker yang ditampilkan pada peta.
# Nilai berasal dari atribut aktual setiap titik.

filter_records = []

for layer_name, gdf in point_layers.items():
    for _, row in gdf.iterrows():
        if row.geometry is None or row.geometry.is_empty:
            continue

        filter_records.append({
            "layer": layer_name,
            "id": str(row.get(
                "_source_id", row.get("_webgis_id", "")
            )),
            "jenis": str(row.get(
                "_jenis_investigasi", "Titik investigasi"
            )),
            "lat": float(row.geometry.y),
            "lon": float(row.geometry.x)
        })

filter_json = json.dumps(
    filter_records, ensure_ascii=False
).replace("</", "<\\/")

filter_html = """
<div id="filter-panel" style="
position:fixed;top:90px;left:15px;z-index:9998;
width:235px;background:white;border:1px solid #888;
border-radius:5px;padding:10px;
font-size:12px;box-shadow:0 1px 5px rgba(0,0,0,0.25);
">
<b>FILTER TITIK</b>
<p style="margin:5px 0;">Jenis investigasi</p>
<select id="filter-type" style="width:100%;padding:5px;">
<option value="ALL">Semua jenis</option>
</select>
<p style="margin:7px 0 5px;">Cari ID</p>
<input id="filter-id" placeholder="Masukkan ID..."
style="width:96%;padding:4px;">
<p id="filter-count" style="margin:7px 0;">
Menyiapkan data...
</p>
<button id="filter-reset" style="padding:4px 8px;">
Reset filter
</button>
<p style="color:#666;font-size:10px;">
Filter bekerja pada marker hasil pemuatan awal.
</p>
</div>
"""

m.get_root().html.add_child(Element(filter_html))

filter_js = """
<script>
(function() {
    const records = __FILTER_DATA__;
    const typeSelect = document.getElementById("filter-type");
    const idInput = document.getElementById("filter-id");
    const countLabel = document.getElementById("filter-count");
    const resetButton = document.getElementById("filter-reset");

    if (!typeSelect || !idInput || !countLabel) return;

    const types = [...new Set(records.map(r => r.jenis))].sort();

    types.forEach(type => {
        const option = document.createElement("option");
        option.value = type;
        option.textContent = type;
        typeSelect.appendChild(option);
    });

    // Kumpulkan marker Folium dari layer yang dibuat.
    const markerRegistry = [];

    function collectLayers(layer) {
        if (!layer || typeof layer.eachLayer !== "function") return;

        layer.eachLayer(function(child) {
            if (child instanceof L.Marker) {
                markerRegistry.push(child);
            } else if (child && typeof child.eachLayer === "function") {
                collectLayers(child);
            }
        });
    }

    // Cari seluruh FeatureGroup pada peta.
    function findMap() {
        for (const key in window) {
            try {
                const value = window[key];
                if (value instanceof L.Map) return value;
            } catch(e) {}
        }
        return null;
    }

    const map = findMap();

    function applyFilter() {
        if (!map) {
            countLabel.textContent =
                "Peta siap; filter marker tidak tersedia.";
            return;
        }

        const selectedType = typeSelect.value;
        const query = idInput.value.trim().toLowerCase();

        let visible = 0;
        let matched = 0;

        // Filter menggunakan koordinat dan atribut ID.
        records.forEach(record => {
            const typeMatch =
                selectedType === "ALL" ||
                record.jenis === selectedType;

            const idMatch =
                !query ||
                record.id.toLowerCase().includes(query);

            if (typeMatch && idMatch) matched++;
        });

        // Marker diatur berdasarkan urutan record.
        // Referensi marker disiapkan pada objek dengan koordinat.
        markerRegistry.forEach(marker => {
            const latlng = marker.getLatLng();

            const record = records.find(r =>
                Math.abs(r.lat - latlng.lat) < 1e-8 &&
                Math.abs(r.lon - latlng.lng) < 1e-8
            );

            if (!record) return;

            const show =
                (selectedType === "ALL" ||
                 record.jenis === selectedType) &&
                (!query || record.id.toLowerCase().includes(query));

            if (show) {
                if (!map.hasLayer(marker)) marker.addTo(map);
                visible++;
            } else {
                if (map.hasLayer(marker)) map.removeLayer(marker);
            }
        });

        countLabel.textContent =
            "Cocok: " + matched + " titik. " +
            "Marker terlihat: " + visible + ".";
    }

    // Beri waktu Leaflet untuk membangun layer.
    setTimeout(function() {
        markerRegistry.length = 0;

        map.eachLayer(function(layer) {
            collectLayers(layer);
        });

        // Hilangkan marker duplikat dari registry.
        const uniqueMarkers = [...new Set(markerRegistry)];
        markerRegistry.length = 0;
        markerRegistry.push(...uniqueMarkers);

        applyFilter();
    }, 300);

    typeSelect.addEventListener("change", applyFilter);
    idInput.addEventListener("input", applyFilter);

    resetButton.addEventListener("click", function() {
        typeSelect.value = "ALL";
        idInput.value = "";
        applyFilter();
    });
})();
</script>
""".replace("__FILTER_DATA__", filter_json)

m.get_root().html.add_child(Element(filter_js))


# ================================================================
# 12. EKSPOR GEOJSON TITIK
# ================================================================

if point_layers:
    combined = gpd.GeoDataFrame(
        pd.concat(
            [gdf.to_crs(WEB_CRS) for gdf in point_layers.values()],
            ignore_index=True
        ),
        geometry="geometry",
        crs=WEB_CRS
    )

    # Simpan gabungan titik untuk pertukaran data.
    geojson_output = OUT_DIR / "investigasi_geoteknik.geojson"
    combined.to_file(
        geojson_output,
        driver="GeoJSON"
    )

    # Ekspor CSV dengan koordinat eksplisit.
    export_df = combined.drop(
        columns=["geometry"], errors="ignore"
    ).copy()

    export_df["Longitude"] = combined.geometry.x
    export_df["Latitude"] = combined.geometry.y

    export_df.to_csv(
        OUT_DIR / "investigasi_geoteknik.csv",
        index=False,
        encoding="utf-8-sig"
    )


# ================================================================
# 13. EKSPOR WEB GIS
# ================================================================

html_output = OUT_DIR / "webgis_investigasi_geoteknik.html"

m.save(str(html_output))

print("\n" + "=" * 65)
print("WEB GIS SELESAI DIBUAT")
print("=" * 65)
print("Peta HTML :", html_output)
print("Laporan QC:", qc_path)

print("\nIsi folder output:")
for file in sorted(OUT_DIR.iterdir()):
    if file.is_file():
        print("-", file.name)

print("\nCara membuka:")
print("1. Buka folder output_webgis.")
print("2. Klik dua kali webgis_investigasi_geoteknik.html.")
print("3. Gunakan browser untuk mengeksplorasi peta.")
````

## Analisis Kritis Latar Belakang dan Tinjauan Pustaka dalam Pengembangan *Research Gap* dan *Novelty* Penelitian Geofisika untuk Publikasi Jurnal Internasional

### R — Role

Bertindak sebagai **research assistant dan akademisi bidang geofisika** yang berpengalaman dalam penulisan artikel ilmiah, systematic/narrative literature review, analisis research gap, pengembangan novelty, dan penyusunan manuskrip untuk jurnal internasional terindeks Scopus. Memahami bidang geofisika terapan, geoteknik, seismologi, geologi, geospatial science, serta mampu mengevaluasi kualitas dan relevansi literatur ilmiah secara kritis.

---

### T — Task

Analisis secara kritis **bagian latar belakang dan tinjauan pustaka** dari manuskrip penelitian yang diberikan. Identifikasi permasalahan ilmiah utama, perkembangan penelitian terdahulu, keterbatasan penelitian sebelumnya, *research gap*, dan potensi *novelty* penelitian.

Jangan hanya meringkas setiap artikel. Lakukan **sintesis antarpenelitian** untuk menunjukkan bagaimana penelitian terdahulu saling berhubungan, apa yang telah diketahui, apa yang masih belum diketahui, dan bagaimana penelitian yang sedang dikembangkan dapat mengisi kesenjangan tersebut.

Setelah analisis dilakukan, berikan rekomendasi terhadap struktur argumentasi latar belakang dan tinjauan pustaka agar memiliki alur ilmiah yang logis:

**konteks umum → permasalahan ilmiah → penelitian terdahulu → keterbatasan → research gap → novelty → tujuan penelitian.**

---

### C — Context

Penelitian yang akan dipublikasikan merupakan penelitian ilmiah di bidang **geofisika/geofisika terapan** yang ditujukan untuk jurnal internasional terindeks Scopus.

Latar belakang tidak boleh hanya berisi informasi umum mengenai topik penelitian. Setiap paragraf harus memiliki fungsi dalam membangun argumentasi ilmiah dan mengarahkan pembaca menuju permasalahan penelitian.

Tinjauan pustaka juga tidak boleh berupa kumpulan ringkasan penelitian terdahulu secara terpisah. Literatur harus disintesis berdasarkan tema, metode, temuan, keterbatasan, atau perkembangan penelitian sehingga menghasilkan argumentasi yang menunjukkan posisi penelitian yang sedang dilakukan.

Analisis harus membedakan antara:

* fakta yang telah didukung penelitian terdahulu;
* interpretasi dari penelitian terdahulu;
* keterbatasan metodologis atau data;
* *research gap*;
* kontribusi penelitian;
* *novelty* yang benar-benar dapat dipertanggungjawabkan.

---

### I — Input

Gunakan informasi yang diberikan pengguna sebagai bahan utama, yang dapat berupa:

1. **Judul penelitian**
2. **Abstrak**
3. **Latar belakang**
4. **Tinjauan pustaka**
5. **Rumusan masalah**
6. **Tujuan penelitian**
7. **Metodologi**
8. **Hasil sementara atau hasil penelitian**
9. **Daftar artikel/jurnal referensi**
10. DOI atau metadata publikasi jika tersedia
11. Topik atau bidang spesifik penelitian

Jika tersedia, gunakan artikel penelitian terdahulu sebagai dasar untuk melakukan perbandingan dan sintesis.

Jangan menganggap informasi yang tidak diberikan pengguna sebagai fakta penelitian. Jika informasi penting tidak tersedia, nyatakan bagian yang masih membutuhkan data atau referensi tambahan.

---

### K — Knowledge & Constraints

Dalam melakukan analisis, gunakan prinsip berikut:

**1. Analisis harus berbasis evidence**

Setiap klaim ilmiah penting harus dapat ditelusuri ke sumber yang relevan. Jangan membuat referensi, DOI, nama penulis, tahun publikasi, hasil penelitian, maupun data statistik yang tidak tersedia.

**2. Bedakan gap dengan sekadar perbedaan penelitian**

Perbedaan lokasi, metode, atau dataset tidak otomatis merupakan *research gap*. Tentukan apakah perbedaan tersebut benar-benar menghasilkan pertanyaan ilmiah yang belum terjawab atau keterbatasan pengetahuan.

**3. Identifikasi jenis research gap**

Jika memungkinkan, klasifikasikan gap sebagai:

* *empirical gap*;
* *methodological gap*;
* *theoretical/conceptual gap*;
* *contextual/geographical gap*;
* *data gap*;
* *knowledge gap*.

Jelaskan dasar identifikasinya.

**4. Novelty harus realistis**

Jangan menyatakan penelitian memiliki novelty hanya karena menggunakan lokasi baru, dataset baru, atau metode yang berbeda. Evaluasi apakah kombinasi data, metode, pendekatan, atau temuan benar-benar memberikan kontribusi ilmiah.

**5. Hindari literature dumping**

Jangan menyusun tinjauan pustaka dalam bentuk:

> Peneliti A menemukan... Peneliti B menemukan... Peneliti C menemukan...

tanpa menjelaskan hubungan dan perbedaan antarpenelitian.

**6. Prioritaskan literatur yang relevan**

Berikan perhatian lebih besar kepada artikel *peer-reviewed* yang relevan dengan topik, metode, wilayah, dan permasalahan penelitian.

**7. Jangan mengarang referensi**

Jika informasi bibliografis tidak tersedia atau tidak dapat diverifikasi, tandai sebagai informasi yang perlu diverifikasi.

**8. Jangan menjanjikan publikasi**

Target Scopus digunakan sebagai orientasi kualitas manuskrip, bukan sebagai jaminan bahwa artikel akan diterima.

**9. Pertahankan integritas ilmiah**

Jangan mengubah hasil penelitian, memanipulasi evidence, atau membuat klaim novelty yang tidak didukung oleh literatur dan hasil penelitian.

---

### O — Output

Berikan hasil analisis secara sistematis dalam urutan berikut:

**A. Diagnosis Latar Belakang**

* fungsi setiap paragraf;
* kekuatan argumentasi;
* kelemahan argumentasi;
* informasi yang masih kurang;
* bagian yang terlalu umum atau tidak relevan.

**B. Pemetaan Literatur**

Buat sintesis berdasarkan:

* penelitian;
* metode;
* data;
* wilayah/konteks;
* temuan utama;
* keterbatasan.

**C. Sintesis Antarpenelitian**

Jelaskan hubungan, persamaan, perbedaan, perkembangan, dan kontradiksi antarpenelitian. Jangan hanya memberikan ringkasan artikel satu per satu.

**D. Identifikasi Research Gap**

Tentukan gap yang paling relevan dan jelaskan:

* apa yang sudah diketahui;
* apa yang belum diketahui;
* bukti yang menunjukkan adanya gap;
* mengapa gap tersebut penting.

**E. Evaluasi Novelty**

Identifikasi potensi novelty penelitian berdasarkan:

* data;
* metode;
* integrasi metode;
* konteks;
* pendekatan analisis;
* temuan;
* kontribusi terhadap pengetahuan.

Bedakan antara **novelty yang benar-benar didukung evidence** dan novelty yang masih berupa klaim.

**F. Research Positioning**

Tunjukkan posisi penelitian pengguna dibandingkan penelitian terdahulu dalam bentuk alur:

**Previous studies → limitation → research gap → proposed approach → expected contribution**

**G. Struktur Latar Belakang yang Direkomendasikan**

Susun kerangka paragraf yang ideal dari paragraf pertama hingga paragraf terakhir.

**H. Struktur Tinjauan Pustaka**

Rekomendasikan pengelompokan literatur berdasarkan tema atau konsep, bukan hanya berdasarkan urutan tahun publikasi.

**I. Draft Perbaikan**

Jika diminta, tuliskan versi revisi latar belakang dan/atau tinjauan pustaka dengan gaya akademik jurnal internasional, mempertahankan substansi dan evidence dari penelitian pengguna.

**J. Final Research Gap & Novelty Statement**

Berikan formulasi singkat dan akademik mengenai:

* *research gap statement*;
* *novelty statement*;
* *scientific contribution statement*.

Jangan membuat klaim yang lebih kuat daripada evidence yang tersedia.

---

### Q — Quality Criteria

Hasil analisis harus memenuhi kriteria berikut:

1. **Evidence-based** — setiap klaim utama memiliki dasar ilmiah.
2. **Critical** — tidak hanya merangkum tetapi membandingkan dan mengevaluasi penelitian.
3. **Coherent** — terdapat hubungan logis antara masalah, literatur, gap, novelty, dan tujuan.
4. **Specific** — research gap tidak ditulis secara generik.
5. **Defensible** — novelty dapat dipertanggungjawabkan berdasarkan literatur dan hasil penelitian.
6. **Non-repetitive** — tidak terjadi pengulangan informasi antarparagraf.
7. **Academic** — menggunakan gaya penulisan ilmiah dan terminologi yang tepat.
8. **Traceable** — sumber setiap klaim dapat ditelusuri.
9. **Publication-oriented** — struktur argumentasi sesuai dengan standar manuskrip jurnal internasional.
10. **Scientifically honest** — tidak membuat klaim berlebihan mengenai novelty, significance, atau kontribusi.

---

### V — Verification

Sebelum memberikan hasil akhir, lakukan pemeriksaan:

**1. Reference verification**

* Apakah referensi benar-benar mendukung klaim?
* Apakah informasi bibliografis tersedia?
* Apakah ada referensi yang tidak dapat diverifikasi?

**2. Gap verification**

* Apakah gap benar-benar berasal dari analisis literatur?
* Apakah gap hanya berupa perbedaan lokasi atau benar-benar merupakan kesenjangan pengetahuan?

**3. Novelty verification**

* Apakah novelty berbeda dari penelitian terdahulu?
* Apakah novelty didukung oleh data dan metode penelitian?
* Apakah klaim novelty terlalu kuat?

**4. Logical-flow verification**

* Apakah latar belakang mengarah secara logis menuju research problem?
* Apakah tinjauan pustaka mengarah menuju research gap?
* Apakah research gap mengarah menuju tujuan penelitian?

**5. Consistency verification**

* Apakah research gap konsisten dengan metode?
* Apakah novelty konsisten dengan hasil?
* Apakah tujuan penelitian menjawab gap yang telah diidentifikasi?



````python

from pathlib import Path
import os
import re
import json
import time
import pandas as pd
import fitz
from tqdm import tqdm
from openai import OpenAI

# ============================================================
# 1. KONFIGURASI
# ============================================================

INPUT_DIR = Path(
    r"D:\RIDWAN\PAPER HSVSR #2\Sumber\Geologi regional"
)

OUTPUT_DIR = INPUT_DIR / "Analisis_Prompt5"
REPORT_DIR = OUTPUT_DIR / "Laporan_Paper"
TEXT_DIR = OUTPUT_DIR / "Teks_Ekstraksi"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
REPORT_DIR.mkdir(parents=True, exist_ok=True)
TEXT_DIR.mkdir(parents=True, exist_ok=True)

EXCEL_PATH = OUTPUT_DIR / "Matriks_Analisis_Literatur.xlsx"
SYNTHESIS_PATH = OUTPUT_DIR / "Sintesis_Lintas_Paper.md"

MODEL = "gpt-5-mini"

# Maksimum teks PDF yang dikirim untuk satu analisis.
# Jika paper sangat panjang, teks akan dipotong.
MAX_CHARS = 100_000

# Jeda singkat antarpaper
REQUEST_DELAY = 1

# ============================================================
# 2. INISIALISASI OPENAI
# ============================================================

api_key = os.getenv("OPENAI_API_KEY")

if not api_key:
    raise RuntimeError(
        "OPENAI_API_KEY belum ditemukan.\n"
        "Atur API key di PowerShell sebelum menjalankan program."
    )

client = OpenAI(api_key=api_key)


# ============================================================
# 3. EKSTRAKSI TEKS PDF
# ============================================================

def extract_pdf(pdf_path):
    pages = []
    page_count = 0

    with fitz.open(pdf_path) as pdf:
        page_count = len(pdf)

        for page_number, page in enumerate(pdf, start=1):
            text = page.get_text("text", sort=True)

            pages.append(
                f"\n\n--- HALAMAN {page_number} ---\n{text}"
            )

    full_text = "\n".join(pages)

    full_text = full_text.replace("\x00", " ")
    full_text = re.sub(
        r"-\s*\n\s*",
        "",
        full_text
    )
    full_text = re.sub(
        r"[ \t]+",
        " ",
        full_text
    )

    return full_text.strip(), page_count


# ============================================================
# 4. FUNGSI MEMANGKAS TEKS SECARA AMAN
# ============================================================

def prepare_text(text):
    """
    Mengutamakan bagian awal dan akhir paper agar abstrak,
    pendahuluan, kesimpulan, serta sebagian referensi tetap
    mungkin terbaca ketika teks harus dipangkas.
    """

    if len(text) <= MAX_CHARS:
        return text

    first_part = int(MAX_CHARS * 0.70)
    last_part = MAX_CHARS - first_part

    return (
        text[:first_part]
        + "\n\n[SEBAGIAN TEKS TENGAH TIDAK DISERTAKAN]\n\n"
        + text[-last_part:]
    )


# ============================================================
# 5. PROMPT ANALISIS INDIVIDUAL PAPER
# ============================================================

SYSTEM_PROMPT = """
Anda adalah asisten analisis literatur ilmiah untuk penelitian
geologi regional, geofisika, geologi teknik, dan kebencanaan.

Analisis hanya berdasarkan teks paper yang diberikan.

ATURAN WAJIB:
1. Jangan mengarang fakta, angka, lokasi, metode, atau sitasi.
2. Jika informasi tidak ditemukan, tulis "Tidak dinyatakan".
3. Bedakan pernyataan eksplisit penulis dengan interpretasi Anda.
4. Jangan menyatakan hubungan sebab-akibat jika paper hanya
   menunjukkan korelasi atau kedekatan spasial.
5. Research gap harus dibedakan menjadi:
   - gap yang dinyatakan eksplisit oleh penulis;
   - gap potensial yang ditafsirkan dari keterbatasan paper.
6. Jangan menyebut suatu gap sebagai gap seluruh bidang hanya
   berdasarkan satu paper.
7. Jangan menganggap kandidat judul atau tahun dari ekstraksi
   otomatis sebagai metadata yang sudah terverifikasi.
8. Jika memungkinkan, sertakan nomor halaman PDF yang mendukung
   temuan. Jangan menebak nomor halaman.
9. Pertahankan istilah teknis sebagaimana digunakan dalam paper.
10. Jawab dalam bahasa Indonesia akademik.
11. Jangan mengarang informasi bibliografi.
12. Jika paper membahas geologi regional, identifikasi
    stratigrafi, litologi, struktur, tektonik, dan lokasi hanya
    sejauh benar-benar tersedia dalam sumber.
"""


def analyze_paper(text, filename):
    user_prompt = f"""
Analisis paper berikut dengan kerangka Prompt 5.

NAMA FILE:
{filename}

TEKS PAPER:
{text}

Kembalikan JSON valid dengan struktur berikut:

{{
  "judul": "",
  "penulis": "",
  "tahun": "",
  "nama_jurnal": "",
  "doi": "",
  "lokasi_penelitian": "",
  "tujuan_penelitian": "",
  "konteks_dan_masalah": "",
  "metode": "",
  "data": "",
  "temuan_utama": "",
  "geologi_regional": "",
  "keterbatasan_eksplisit": "",
  "research_gap_eksplisit": "",
  "research_gap_potensial": "",
  "kontribusi_dan_kebaruan_yang_diklaim": "",
  "relevansi_terhadap_manuskrip": "",
  "bukti_dan_halaman": "",
  "informasi_yang_tidak_ditemukan": "",
  "catatan_kualitas_ekstraksi": ""
}}

Ketentuan pengisian:
- Gunakan paragraf ringkas tetapi informatif.
- Metode dan data harus dibedakan.
- Jangan menyamakan keterbatasan penelitian dengan research gap
  yang sudah terbukti.
- Bagian geologi regional hanya berisi informasi yang relevan
  dan benar-benar ditemukan.
- Pada bukti_dan_halaman, cantumkan klaim utama beserta halaman
  PDF jika dapat dikenali dari penanda HALAMAN.
- Relevansi terhadap manuskrip adalah evaluasi awal, bukan
  bukti bahwa paper secara langsung mendukung semua argumen.
- Jika judul, penulis, tahun, atau DOI tidak dapat dipastikan,
  tandai bahwa metadata perlu diverifikasi.
"""

    response = client.responses.create(
        model=MODEL,
        instructions=SYSTEM_PROMPT,
        input=user_prompt,
        text={"format": {"type": "json_object"}}
    )

    return json.loads(response.output_text)


# ============================================================
# 6. MENYUSUN LAPORAN MARKDOWN
# ============================================================

def save_report(filename, analysis):
    report_path = REPORT_DIR / (
        re.sub(r'[<>:"/\\|?*]', "_", filename)
        + ".md"
    )

    labels = {
        "judul": "Judul",
        "penulis": "Penulis",
        "tahun": "Tahun",
        "nama_jurnal": "Nama jurnal",
        "doi": "DOI",
        "lokasi_penelitian": "Lokasi penelitian",
        "tujuan_penelitian": "Tujuan penelitian",
        "konteks_dan_masalah": "Konteks dan masalah penelitian",
        "metode": "Metode penelitian",
        "data": "Data penelitian",
        "temuan_utama": "Temuan utama",
        "geologi_regional": "Informasi geologi regional",
        "keterbatasan_eksplisit": "Keterbatasan eksplisit",
        "research_gap_eksplisit": "Research gap eksplisit",
        "research_gap_potensial": "Research gap potensial",
        "kontribusi_dan_kebaruan_yang_diklaim":
            "Kontribusi dan kebaruan yang diklaim",
        "relevansi_terhadap_manuskrip":
            "Relevansi terhadap manuskrip",
        "bukti_dan_halaman": "Bukti dan halaman",
        "informasi_yang_tidak_ditemukan":
            "Informasi yang tidak ditemukan",
        "catatan_kualitas_ekstraksi":
            "Catatan kualitas ekstraksi"
    }

    content = [
        "# Analisis Paper: Prompt 5",
        "",
        f"**Nama file:** {filename}",
        "",
        "> Laporan ini merupakan analisis berbantuan AI atas teks "
        "PDF yang berhasil diekstrak. Verifikasi manual tetap "
        "diperlukan sebelum digunakan dalam manuskrip.",
        ""
    ]

    for key, label in labels.items():
        content.extend([
            f"## {label}",
            str(analysis.get(key, "Tidak dinyatakan")),
            ""
        ])

    report_path.write_text(
        "\n".join(content),
        encoding="utf-8"
    )

    return report_path


# ============================================================
# 7. ANALISIS LINTAS PAPER
# ============================================================

def synthesize_papers(summaries):
    """
    Membandingkan temuan lintas paper.
    Sintesis dibatasi pada hasil analisis individual.
    """

    input_data = json.dumps(
        summaries,
        ensure_ascii=False,
        indent=2
    )

    prompt = f"""
Buat sintesis lintas paper berdasarkan data analisis berikut.

DATA ANALISIS:
{input_data}

Susun laporan dalam bahasa Indonesia dengan bagian:

1. Gambaran umum korpus literatur.
2. Tema dan fokus penelitian yang dominan.
3. Perbandingan lokasi, metode, dan jenis data.
4. Kesepakatan dan perbedaan temuan antar-paper.
5. Kontribusi setiap paper terhadap geologi regional.
6. Keterbatasan yang berulang.
7. Research gap eksplisit yang disebutkan penulis.
8. Research gap potensial hasil sintesis.
9. Implikasi untuk penyusunan latar belakang manuskrip.
10. Referensi mana yang perlu diperiksa lebih lanjut.

ATURAN:
- Gunakan nama file untuk menelusuri asal setiap klaim.
- Jangan menganggap informasi yang tidak disebutkan sebagai
  bukti bahwa topik tersebut belum pernah diteliti.
- Jangan menyimpulkan novelty manuskrip tanpa mengetahui
  pertanyaan dan isi manuskrip yang akan diteliti.
- Jika bukti antarpaper tidak cukup, nyatakan keterbatasannya.
- Jangan membuat daftar pustaka atau DOI yang tidak tersedia.
- Bedakan temuan sumber dari interpretasi sintesis.
"""

    response = client.responses.create(
        model=MODEL,
        instructions=SYSTEM_PROMPT,
        input=prompt
    )

    SYNTHESIS_PATH.write_text(
        response.output_text,
        encoding="utf-8"
    )


# ============================================================
# 8. PROGRAM UTAMA
# ============================================================

def main():
    pdf_files = sorted(
        path for path in INPUT_DIR.rglob("*.pdf")
        if OUTPUT_DIR not in path.parents
    )

    if not pdf_files:
        print("Tidak ditemukan file PDF di folder sumber.")
        return

    print(f"Jumlah PDF ditemukan: {len(pdf_files)}")

    rows = []
    synthesis_summaries = []

    for pdf_path in tqdm(pdf_files, desc="Analisis paper"):
        relative_name = str(
            pdf_path.relative_to(INPUT_DIR)
        )

        print(f"\nMemproses: {relative_name}")

        try:
            # A. Ekstraksi PDF
            text, page_count = extract_pdf(pdf_path)

            if len(text.strip()) < 200:
                raise ValueError(
                    "Teks sangat sedikit. PDF mungkin hasil "
                    "scan dan memerlukan OCR."
                )

            # Simpan teks ekstraksi
            safe_name = re.sub(
                r'[<>:"/\\|?*]',
                "_",
                relative_name
            )

            text_path = TEXT_DIR / (
                Path(safe_name).with_suffix(".txt")
            )
            text_path.parent.mkdir(
                parents=True,
                exist_ok=True
            )

            text_path.write_text(
                text,
                encoding="utf-8"
            )

            # B. Analisis AI
            analysis_text = prepare_text(text)
            analysis = analyze_paper(
                analysis_text,
                relative_name
            )

            # Simpan laporan detail
            report_path = save_report(
                Path(relative_name).stem,
                analysis
            )

            # C. Baris matriks Excel
            row = {
                "Nama File": relative_name,
                "Jumlah Halaman PDF": page_count,
                "Status": "Berhasil",
                **analysis,
                "File Teks": str(text_path),
                "File Laporan": str(report_path)
            }

            rows.append(row)

            synthesis_summaries.append({
                "nama_file": relative_name,
                "judul": analysis.get("judul", ""),
                "penulis": analysis.get("penulis", ""),
                "tahun": analysis.get("tahun", ""),
                "lokasi_penelitian":
                    analysis.get("lokasi_penelitian", ""),
                "tujuan_penelitian":
                    analysis.get("tujuan_penelitian", ""),
                "metode": analysis.get("metode", ""),
                "temuan_utama":
                    analysis.get("temuan_utama", ""),
                "geologi_regional":
                    analysis.get("geologi_regional", ""),
                "keterbatasan":
                    analysis.get("keterbatasan_eksplisit", ""),
                "gap_eksplisit":
                    analysis.get("research_gap_eksplisit", ""),
                "gap_potensial":
                    analysis.get("research_gap_potensial", ""),
                "bukti_dan_halaman":
                    analysis.get("bukti_dan_halaman", "")
            })

            print("Analisis berhasil.")

        except Exception as error:
            print(f"GAGAL: {error}")

            rows.append({
                "Nama File": relative_name,
                "Status": f"Gagal: {error}"
            })

        time.sleep(REQUEST_DELAY)

    # D. Simpan matriks Excel
    df = pd.DataFrame(rows)

    with pd.ExcelWriter(
        EXCEL_PATH,
        engine="openpyxl"
    ) as writer:
        df.to_excel(
            writer,
            sheet_name="Matriks Literatur",
            index=False
        )

        for sheet in writer.sheets.values():
            sheet.freeze_panes = "A2"
            sheet.auto_filter.ref = sheet.dimensions

            for cells in sheet.columns:
                letter = cells[0].column_letter
                sheet.column_dimensions[letter].width = 25

    # E. Sintesis lintas paper
    if synthesis_summaries:
        try:
            print("\nMenyusun sintesis lintas paper...")
            synthesize_papers(synthesis_summaries)
            print(f"Sintesis tersimpan: {SYNTHESIS_PATH}")

        except Exception as error:
            print(
                "Sintesis lintas paper gagal dibuat: "
                f"{error}"
            )

    print("\n" + "=" * 55)
    print("PROSES SELESAI")
    print("=" * 55)
    print(f"PDF ditemukan : {len(pdf_files)}")
    print(f"Excel         : {EXCEL_PATH}")
    print(f"Laporan       : {REPORT_DIR}")
    print(f"Teks PDF      : {TEXT_DIR}")
    print(f"Sintesis      : {SYNTHESIS_PATH}")


if __name__ == "__main__":
    main()
````
